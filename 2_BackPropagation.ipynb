```python
# ================================================================
# EXERCISE 2
# BACKPROPAGATION NEURAL NETWORK PREDICTION AND FEATURE ANALYSIS
# Breast Cancer Wisconsin (Diagnostic) Dataset
# ================================================================

# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

from itertools import combinations

print("Libraries imported successfully!")


# ================================================================
# 2. LOAD DATASET
# ================================================================

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00267/data_banknote_authentication.txt"

# The URL above is for Exercise 1, so for Exercise 2 we use the
# Breast Cancer Wisconsin Diagnostic dataset from UCI.

data_url = (
    "https://archive.ics.uci.edu/ml/machine-learning-databases/"
    "breast-cancer-wisconsin/wdbc.data"
)

# Feature names
feature_names = [
    "radius_mean",
    "texture_mean",
    "perimeter_mean",
    "area_mean",
    "smoothness_mean",
    "compactness_mean",
    "concavity_mean",
    "concave_points_mean",
    "symmetry_mean",
    "fractal_dimension_mean",

    "radius_se",
    "texture_se",
    "perimeter_se",
    "area_se",
    "smoothness_se",
    "compactness_se",
    "concavity_se",
    "concave_points_se",
    "symmetry_se",
    "fractal_dimension_se",

    "radius_worst",
    "texture_worst",
    "perimeter_worst",
    "area_worst",
    "smoothness_worst",
    "compactness_worst",
    "concavity_worst",
    "concave_points_worst",
    "symmetry_worst",
    "fractal_dimension_worst"
]

columns = ["ID", "Diagnosis"] + feature_names

df = pd.read_csv(
    data_url,
    header=None,
    names=columns
)

print("\nDataset loaded successfully!")

print("Number of samples:", len(df))
print("Number of features:", len(feature_names))


# ================================================================
# 3. ENCODE TARGET
# ================================================================

# B = Benign = 0
# M = Malignant = 1

df["Diagnosis"] = df["Diagnosis"].map({
    "B": 0,
    "M": 1
})

print("\nTarget encoding:")
print("Benign = 0")
print("Malignant = 1")

print("\nClass distribution:")
print(df["Diagnosis"].value_counts())


# ================================================================
# 4. PREPARE X AND y
# ================================================================

X = df[feature_names].values.astype(float)

y = df["Diagnosis"].values.astype(int)

print("\nX shape:", X.shape)
print("y shape:", y.shape)


# ================================================================
# 5. NORMALIZE FEATURES
# ================================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("\nFeatures normalized successfully!")


# ================================================================
# 6. TRAIN / TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# ================================================================
# 7. SIGMOID FUNCTION
# ================================================================

def sigmoid(x):

    # Prevent overflow
    x = np.clip(x, -500, 500)

    return 1 / (1 + np.exp(-x))


# ================================================================
# 8. CUSTOM NEURAL NETWORK
# ================================================================

class NeuralNetwork:

    def __init__(
        self,
        input_size,
        hidden_layers=(10,),
        learning_rate=0.01,
        max_epochs=500,
        random_state=42
    ):

        self.input_size = input_size
        self.hidden_layers = hidden_layers
        self.learning_rate = learning_rate
        self.max_epochs = max_epochs
        self.random_state = random_state

        self.weights = []
        self.biases = []

        self.loss_history = []

        self.epochs = 0
        self.converged = False

        self._initialize_weights()


    # ------------------------------------------------------------
    # Initialize weights
    # ------------------------------------------------------------

    def _initialize_weights(self):

        rng = np.random.default_rng(self.random_state)

        layer_sizes = [
            self.input_size
        ] + list(self.hidden_layers) + [1]

        for i in range(len(layer_sizes) - 1):

            input_nodes = layer_sizes[i]
            output_nodes = layer_sizes[i + 1]

            # Xavier-style initialization
            limit = np.sqrt(
                6 / (input_nodes + output_nodes)
            )

            W = rng.uniform(
                -limit,
                limit,
                size=(input_nodes, output_nodes)
            )

            b = np.zeros((1, output_nodes))

            self.weights.append(W)
            self.biases.append(b)


    # ------------------------------------------------------------
    # Forward propagation
    # ------------------------------------------------------------

    def forward(self, X):

        activations = [X]

        current = X

        for W, b in zip(self.weights, self.biases):

            z = np.dot(current, W) + b

            current = sigmoid(z)

            activations.append(current)

        return activations


    # ------------------------------------------------------------
    # Train using backpropagation
    # ------------------------------------------------------------

    def fit(self, X, y):

        y = y.reshape(-1, 1)

        self.loss_history = []

        self.converged = False

        for epoch in range(self.max_epochs):

            # -----------------------------
            # Forward propagation
            # -----------------------------

            activations = self.forward(X)

            output = activations[-1]

            # -----------------------------
            # Binary cross entropy loss
            # -----------------------------

            eps = 1e-10

            loss = -np.mean(
                y * np.log(output + eps)
                +
                (1 - y) * np.log(1 - output + eps)
            )

            self.loss_history.append(loss)

            # -----------------------------
            # Backpropagation
            # -----------------------------

            gradients_W = [
                np.zeros_like(W)
                for W in self.weights
            ]

            gradients_b = [
                np.zeros_like(b)
                for b in self.biases
            ]

            # Output layer error
            delta = output - y

            # Move backward through layers
            for layer in reversed(range(len(self.weights))):

                gradients_W[layer] = (
                    np.dot(
                        activations[layer].T,
                        delta
                    )
                    / len(X)
                )

                gradients_b[layer] = (
                    np.mean(
                        delta,
                        axis=0,
                        keepdims=True
                    )
                )

                if layer > 0:

                    previous_activation = activations[layer]

                    delta = np.dot(
                        delta,
                        self.weights[layer].T
                    ) * previous_activation * (
                        1 - previous_activation
                    )

            # -----------------------------
            # Update weights and biases
            # -----------------------------

            for layer in range(len(self.weights)):

                self.weights[layer] -= (
                    self.learning_rate *
                    gradients_W[layer]
                )

                self.biases[layer] -= (
                    self.learning_rate *
                    gradients_b[layer]
                )

            self.epochs = epoch + 1

            # Simple convergence condition
            if loss < 0.02:

                self.converged = True
                break

        return self


    # ------------------------------------------------------------
    # Predict probability/output
    # ------------------------------------------------------------

    def predict_proba(self, X):

        activations = self.forward(X)

        return activations[-1].flatten()


    # ------------------------------------------------------------
    # Predict class
    # ------------------------------------------------------------

    def predict(self, X):

        probabilities = self.predict_proba(X)

        return (probabilities >= 0.5).astype(int)


    # ------------------------------------------------------------
    # Get hidden neuron activations
    # ------------------------------------------------------------

    def hidden_activations(self, X):

        activations = self.forward(X)

        # Everything except input and output
        return activations[1:-1]


# ================================================================
# 9. CREATE MAIN BPNN
# ================================================================

bpnn = NeuralNetwork(
    input_size=30,
    hidden_layers=(10,),
    learning_rate=0.05,
    max_epochs=500,
    random_state=42
)

bpnn.fit(
    X_train,
    y_train
)

print("\nBPNN trained successfully!")

print("Epochs:", bpnn.epochs)
print("Converged:", bpnn.converged)


# ================================================================
# 10. MAIN BPNN PREDICTIONS
# ================================================================

bpnn_outputs = bpnn.predict_proba(X_test)

bpnn_predictions = bpnn.predict(X_test)

bpnn_accuracy = accuracy_score(
    y_test,
    bpnn_predictions
)

print("\nBPNN Test Accuracy:")
print(round(bpnn_accuracy * 100, 2), "%")


# ================================================================
# 11. HELPER FUNCTIONS
# ================================================================

def confidence_category(output):

    distance = abs(output - 0.5)

    if distance < 0.10:
        return "Low Confidence"

    elif distance < 0.25:
        return "Moderate Confidence"

    else:
        return "High Confidence"


def get_class_name(value):

    if value == 0:
        return "Benign"

    return "Malignant"


def print_model_summary(model, name):

    predictions = model.predict(X_test)

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    print("\n----------------------------------------")
    print(name)
    print("----------------------------------------")

    print("Architecture:")

    layer_sizes = [
        model.input_size
    ] + list(model.hidden_layers) + [1]

    print(
        " → ".join(
            str(x)
            for x in layer_sizes
        )
    )

    print("Epochs:", model.epochs)

    print("Converged:", model.converged)

    print(
        "Accuracy:",
        round(accuracy * 100, 2),
        "%"
    )


# ================================================================
# OPTION 1
# TUMOR CLASSIFICATION USING BPNN
# ================================================================

def option_1():

    print("\n")
    print("=" * 70)
    print("1. TUMOR CLASSIFICATION USING BPNN")
    print("=" * 70)

    results = pd.DataFrame({

        "Sample": np.arange(
            1,
            len(y_test) + 1
        ),

        "Actual Class": [
            get_class_name(x)
            for x in y_test
        ],

        "Predicted Class": [
            get_class_name(x)
            for x in bpnn_predictions
        ],

        "Network Output": np.round(
            bpnn_outputs,
            4
        )
    })

    display(results.head(30))

    print(
        "\nClassification Accuracy:",
        round(bpnn_accuracy * 100, 2),
        "%"
    )

    print("\nConfusion Matrix:")

    print(
        confusion_matrix(
            y_test,
            bpnn_predictions
        )
    )

    print("\nClassification Report:")

    print(
        classification_report(
            y_test,
            bpnn_predictions,
            target_names=[
                "Benign",
                "Malignant"
            ]
        )
    )

    # Training loss graph

    plt.figure(figsize=(8, 5))

    plt.plot(
        bpnn.loss_history
    )

    plt.xlabel("Epoch")

    plt.ylabel("Loss")

    plt.title(
        "BPNN Training Loss"
    )

    plt.grid(True)

    plt.show()


# ================================================================
# OPTION 2
# PREDICTION CONFIDENCE ANALYZER
# ================================================================

def option_2():

    print("\n")
    print("=" * 70)
    print("2. PREDICTION CONFIDENCE ANALYZER")
    print("=" * 70)

    confidence_results = pd.DataFrame({

        "Sample": np.arange(
            1,
            len(y_test) + 1
        ),

        "Network Output": np.round(
            bpnn_outputs,
            4
        ),

        "Predicted Class": [
            get_class_name(x)
            for x in bpnn_predictions
        ],

        "Confidence": [
            confidence_category(x)
            for x in bpnn_outputs
        ]
    })

    display(
        confidence_results
    )

    print("\nConfidence interpretation:")

    print(
        "Low      → output close to 0.5"
    )

    print(
        "Moderate → output somewhat away from 0.5"
    )

    print(
        "High     → output far from 0.5"
    )


# ================================================================
# OPTION 3
# FEATURE CONTRIBUTION ANALYZER
# ================================================================

def option_3():

    print("\n")
    print("=" * 70)
    print("3. FEATURE CONTRIBUTION ANALYZER")
    print("=" * 70)

    sample_number = int(
        input(
            f"Enter test sample number (1-{len(X_test)}): "
        )
    )

    if sample_number < 1 or sample_number > len(X_test):

        print("Invalid sample number.")

        return

    index = sample_number - 1

    original_sample = X_test[
        index
    ].copy()

    original_output = bpnn.predict_proba(
        original_sample.reshape(1, -1)
    )[0]

    contributions = []

    for feature_index in range(30):

        modified_sample = (
            original_sample.copy()
        )

        # Since the features are standardized,
        # zero means approximately the dataset mean.
        modified_sample[
            feature_index
        ] = 0

        modified_output = bpnn.predict_proba(
            modified_sample.reshape(1, -1)
        )[0]

        change = abs(
            original_output -
            modified_output
        )

        contributions.append({

            "Feature": feature_names[
                feature_index
            ],

            "Original Output": round(
                original_output,
                6
            ),

            "Modified Output": round(
                modified_output,
                6
            ),

            "Output Change": round(
                change,
                6
            )
        })

    contribution_df = pd.DataFrame(
        contributions
    )

    contribution_df = contribution_df.sort_values(
        by="Output Change",
        ascending=False
    )

    print(
        "\nOriginal Network Output:",
        round(original_output, 6)
    )

    print(
        "\nFeature influence ranking:"
    )

    display(
        contribution_df
    )

    print(
        "\nTop 5 most influential features:"
    )

    display(
        contribution_df.head(5)
    )


# ================================================================
# OPTION 4
# HIDDEN NEURON RESPONSE ANALYZER
# ================================================================

def option_4():

    print("\n")
    print("=" * 70)
    print("4. HIDDEN NEURON RESPONSE ANALYZER")
    print("=" * 70)

    hidden_layers = bpnn.hidden_activations(
        X_test
    )

    # Our BPNN has one hidden layer
    hidden_matrix = hidden_layers[0]

    hidden_columns = [
        f"Neuron_{i + 1}"
        for i in range(
            hidden_matrix.shape[1]
        )
    ]

    hidden_df = pd.DataFrame(
        hidden_matrix,
        columns=hidden_columns
    )

    print(
        "\nHidden Neuron Activation Matrix:"
    )

    display(
        hidden_df.head(20)
    )

    # Separate classes

    benign_mask = (
        y_test == 0
    )

    malignant_mask = (
        y_test == 1
    )

    benign_average = hidden_matrix[
        benign_mask
    ].mean(axis=0)

    malignant_average = hidden_matrix[
        malignant_mask
    ].mean(axis=0)

    analysis_df = pd.DataFrame({

        "Neuron": hidden_columns,

        "Average Activation - Benign":
            np.round(
                benign_average,
                4
            ),

        "Average Activation - Malignant":
            np.round(
                malignant_average,
                4
            )
    })

    print(
        "\nAverage hidden-neuron activation:"
    )

    display(
        analysis_df
    )

    best_benign = np.argmax(
        benign_average
    )

    best_malignant = np.argmax(
        malignant_average
    )

    print(
        "\nHighest average activation for Benign:",
        f"Neuron_{best_benign + 1}"
    )

    print(
        "Activation:",
        round(
            benign_average[
                best_benign
            ],
            4
        )
    )

    print(
        "\nHighest average activation for Malignant:",
        f"Neuron_{best_malignant + 1}"
    )

    print(
        "Activation:",
        round(
            malignant_average[
                best_malignant
            ],
            4
        )
    )


# ================================================================
# OPTION 5
# INPUT FEATURE MASKING CHALLENGE
# ================================================================

def option_5():

    print("\n")
    print("=" * 70)
    print("5. INPUT FEATURE MASKING CHALLENGE")
    print("=" * 70)

    original_predictions = bpnn.predict(
        X_test
    )

    original_accuracy = accuracy_score(
        y_test,
        original_predictions
    )

    results = []

    for feature_index in range(30):

        masked_X = X_test.copy()

        # Zero in standardized space ≈ dataset mean
        masked_X[
            :,
            feature_index
        ] = 0

        masked_predictions = bpnn.predict(
            masked_X
        )

        masked_accuracy = accuracy_score(
            y_test,
            masked_predictions
        )

        accuracy_drop = (
            original_accuracy -
            masked_accuracy
        ) * 100

        results.append({

            "Masked Feature":
                feature_names[
                    feature_index
                ],

            "Original Accuracy (%)":
                round(
                    original_accuracy * 100,
                    2
                ),

            "Masked Accuracy (%)":
                round(
                    masked_accuracy * 100,
                    2
                ),

            "Accuracy Drop (%)":
                round(
                    accuracy_drop,
                    2
                )
        })

    masking_df = pd.DataFrame(
        results
    )

    masking_df = masking_df.sort_values(
        by="Accuracy Drop (%)",
        ascending=False
    )

    print(
        "\nFeature masking results:"
    )

    display(
        masking_df
    )

    print(
        "\nFeature causing maximum accuracy drop:"
    )

    display(
        masking_df.head(1)
    )


# ================================================================
# OPTION 6
# MISCLASSIFICATION CONFIDENCE ANALYZER
# ================================================================

def option_6():

    print("\n")
    print("=" * 70)
    print("6. MISCLASSIFICATION CONFIDENCE ANALYZER")
    print("=" * 70)

    incorrect_indices = np.where(
        bpnn_predictions != y_test
    )[0]

    if len(incorrect_indices) == 0:

        print(
            "\nNo misclassified samples!"
        )

        return

    results = []

    for index in incorrect_indices:

        output = bpnn_outputs[
            index
        ]

        results.append({

            "Sample":
                index + 1,

            "Actual Class":
                get_class_name(
                    y_test[index]
                ),

            "Predicted Class":
                get_class_name(
                    bpnn_predictions[index]
                ),

            "Sigmoid Output":
                round(
                    output,
                    6
                ),

            "Confidence":
                confidence_category(
                    output
                ),

            "Distance from Threshold":
                round(
                    abs(
                        output - 0.5
                    ),
                    6
                )
        })

    misclassification_df = pd.DataFrame(
        results
    )

    print(
        "\nMisclassified samples:"
    )

    display(
        misclassification_df
    )

    # Confidently wrong
    confident_wrong = (
        misclassification_df[
            misclassification_df[
                "Distance from Threshold"
            ] >= 0.25
        ]
    )

    print(
        "\nConfidently wrong predictions:"
    )

    if len(confident_wrong) == 0:

        print(
            "No confidently wrong predictions "
            "under the defined threshold."
        )

    else:

        display(
            confident_wrong
        )


# ================================================================
# OPTION 7
# BPNN vs MLP vs SHLNN
# ================================================================

def option_7():

    print("\n")
    print("=" * 70)
    print("7. BPNN - MLP - SHLNN COMPARATIVE ANALYSIS")
    print("=" * 70)

    # ------------------------------------------------------------
    # BPNN
    # One hidden layer, 10 neurons
    # ------------------------------------------------------------

    model_bpnn = NeuralNetwork(
        input_size=30,
        hidden_layers=(10,),
        learning_rate=0.05,
        max_epochs=500,
        random_state=42
    )

    model_bpnn.fit(
        X_train,
        y_train
    )

    # ------------------------------------------------------------
    # SHLNN
    # One hidden layer, 20 neurons
    # ------------------------------------------------------------

    model_shlnn = NeuralNetwork(
        input_size=30,
        hidden_layers=(20,),
        learning_rate=0.05,
        max_epochs=500,
        random_state=42
    )

    model_shlnn.fit(
        X_train,
        y_train
    )

    # ------------------------------------------------------------
    # MLP
    # Two hidden layers
    # ------------------------------------------------------------

    model_mlp = NeuralNetwork(
        input_size=30,
        hidden_layers=(16, 8),
        learning_rate=0.05,
        max_epochs=500,
        random_state=42
    )

    model_mlp.fit(
        X_train,
        y_train
    )

    # ------------------------------------------------------------
    # Predictions
    # ------------------------------------------------------------

    pred_bpnn = model_bpnn.predict(
        X_test
    )

    pred_shlnn = model_shlnn.predict(
        X_test
    )

    pred_mlp = model_mlp.predict(
        X_test
    )

    # ------------------------------------------------------------
    # Accuracy
    # ------------------------------------------------------------

    acc_bpnn = accuracy_score(
        y_test,
        pred_bpnn
    )

    acc_shlnn = accuracy_score(
        y_test,
        pred_shlnn
    )

    acc_mlp = accuracy_score(
        y_test,
        pred_mlp
    )

    # ------------------------------------------------------------
    # Comparative table
    # ------------------------------------------------------------

    comparison = pd.DataFrame({

        "Model": [
            "BPNN",
            "SHLNN",
            "MLP"
        ],

        "Architecture": [
            "30 → 10 → 1",
            "30 → 20 → 1",
            "30 → 16 → 8 → 1"
        ],

        "Epochs": [
            model_bpnn.epochs,
            model_shlnn.epochs,
            model_mlp.epochs
        ],

        "Accuracy (%)": [
            round(
                acc_bpnn * 100,
                2
            ),

            round(
                acc_shlnn * 100,
                2
            ),

            round(
                acc_mlp * 100,
                2
            )
        ],

        "Converged": [
            model_bpnn.converged,
            model_shlnn.converged,
            model_mlp.converged
        ]
    })

    display(
        comparison
    )

    # ------------------------------------------------------------
    # Classification reports
    # ------------------------------------------------------------

    print("\nBPNN Classification Report:")

    print(
        classification_report(
            y_test,
            pred_bpnn,
            target_names=[
                "Benign",
                "Malignant"
            ]
        )
    )

    print("\nSHLNN Classification Report:")

    print(
        classification_report(
            y_test,
            pred_shlnn,
            target_names=[
                "Benign",
                "Malignant"
            ]
        )
    )

    print("\nMLP Classification Report:")

    print(
        classification_report(
            y_test,
            pred_mlp,
            target_names=[
                "Benign",
                "Malignant"
            ]
        )
    )

    # ------------------------------------------------------------
    # Loss comparison
    # ------------------------------------------------------------

    plt.figure(figsize=(9, 5))

    plt.plot(
        model_bpnn.loss_history,
        label="BPNN"
    )

    plt.plot(
        model_shlnn.loss_history,
        label="SHLNN"
    )

    plt.plot(
        model_mlp.loss_history,
        label="MLP"
    )

    plt.xlabel("Epoch")

    plt.ylabel("Loss")

    plt.title(
        "BPNN vs SHLNN vs MLP Training Loss"
    )

    plt.legend()

    plt.grid(True)

    plt.show()


# ================================================================
# OPTION 8
# EXIT
# ================================================================


# ================================================================
# MENU
# ================================================================

def show_menu():

    while True:

        print("\n")

        print("=" * 70)

        print(
            " BREAST CANCER BPNN / MLP / SHLNN ANALYSIS"
        )

        print("=" * 70)

        print(
            "1. Tumor Classification using BPNN"
        )

        print(
            "2. Prediction Confidence Analyzer"
        )

        print(
            "3. Feature Contribution Analyzer"
        )

        print(
            "4. Hidden Neuron Response Analyzer"
        )

        print(
            "5. Input Feature Masking Challenge"
        )

        print(
            "6. Misclassification Confidence Analyzer"
        )

        print(
            "7. BPNN-MLP-SHLNN Comparative Analysis"
        )

        print(
            "8. Exit"
        )

        print("=" * 70)

        choice = input(
            "Enter your choice: "
        )

        # --------------------------------------------------------
        # OPTION 1
        # --------------------------------------------------------

        if choice == "1":

            option_1()

        # --------------------------------------------------------
        # OPTION 2
        # --------------------------------------------------------

        elif choice == "2":

            option_2()

        # --------------------------------------------------------
        # OPTION 3
        # --------------------------------------------------------

        elif choice == "3":

            option_3()

        # --------------------------------------------------------
        # OPTION 4
        # --------------------------------------------------------

        elif choice == "4":

            option_4()

        # --------------------------------------------------------
        # OPTION 5
        # --------------------------------------------------------

        elif choice == "5":

            option_5()

        # --------------------------------------------------------
        # OPTION 6
        # --------------------------------------------------------

        elif choice == "6":

            option_6()

        # --------------------------------------------------------
        # OPTION 7
        # --------------------------------------------------------

        elif choice == "7":

            option_7()

        # --------------------------------------------------------
        # OPTION 8
        # --------------------------------------------------------

        elif choice == "8":

            print(
                "\nThank you for using the "
                "Breast Cancer Neural Network System!"
            )

            print(
                "Program terminated."
            )

            break

        # --------------------------------------------------------
        # INVALID
        # --------------------------------------------------------

        else:

            print(
                "\nInvalid choice!"
            )

            print(
                "Please enter a number from 1 to 8."
            )


# ================================================================
# START PROGRAM
# ================================================================

print("\n")
print("=" * 70)
print(
    " BREAST CANCER NEURAL NETWORK SYSTEM READY"
)
print("=" * 70)

print(
    "\nBPNN Architecture: 30 → 10 → 1"
)

print(
    "SHLNN Architecture: 30 → 20 → 1"
)

print(
    "MLP Architecture: 30 → 16 → 8 → 1"
)

print(
    "\nTarget Encoding:"
)

print(
    "Benign = 0"
)

print(
    "Malignant = 1"
)

print(
    "\nTraining samples:",
    len(X_train)
)

print(
    "Testing samples:",
    len(X_test)
)

print(
    "\nInitial BPNN accuracy:",
    round(
        bpnn_accuracy * 100,
        2
    ),
    "%"
)

print(
    "\nRun the menu below."
)


# Start menu
show_menu()
```
