```python
# ============================================================
# EXERCISE 3
# Feedforward Neural Network Diagnostic System
# Detecting Learning Rate, Gradient and Overfitting Failures
# Dataset: Fashion-MNIST
# ============================================================

# =========================
# 1. IMPORT LIBRARIES
# =========================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import SGD, Adam
from tensorflow.keras.losses import SparseCategoricalCrossentropy

from sklearn.model_selection import train_test_split

import warnings
warnings.filterwarnings("ignore")


# =========================
# 2. RANDOM SEED
# =========================

np.random.seed(42)
tf.random.set_seed(42)


# =========================
# 3. CONFIGURATION
# =========================

# Set these to None if you want to use the full dataset.
TRAIN_SAMPLES = 12000
VAL_SAMPLES = 2000
TEST_SAMPLES = 3000

BASELINE_EPOCHS = 10
LR_EPOCHS = 8
GRADIENT_EPOCHS = 5
OVERFIT_EPOCHS = 20

BATCH_SIZE = 128


# =========================
# 4. FASHION-MNIST CLASSES
# =========================

CLASS_NAMES = [
    "T-shirt/top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle boot"
]


# =========================
# 5. LOAD DATASET
# =========================

print("=" * 70)
print("LOADING FASHION-MNIST DATASET")
print("=" * 70)

(X_full, y_full), (X_test_full, y_test_full) = tf.keras.datasets.fashion_mnist.load_data()

print("Original training data:", X_full.shape)
print("Original test data:", X_test_full.shape)


# =========================
# 6. NORMALIZE PIXELS
# =========================

X_full = X_full.astype("float32") / 255.0
X_test_full = X_test_full.astype("float32") / 255.0


# =========================
# 7. TRAIN / VALIDATION SPLIT
# =========================

if TRAIN_SAMPLES is not None:

    X_train, _, y_train, _ = train_test_split(
        X_full,
        y_full,
        train_size=TRAIN_SAMPLES,
        stratify=y_full,
        random_state=42
    )

else:

    X_train = X_full
    y_train = y_full


if VAL_SAMPLES is not None:

    X_val, _, y_val, _ = train_test_split(
        X_full,
        y_full,
        train_size=VAL_SAMPLES,
        stratify=y_full,
        random_state=123
    )

else:

    X_val = X_full
    y_val = y_full


# =========================
# 8. TEST SUBSET
# =========================

if TEST_SAMPLES is not None:

    X_test, _, y_test, _ = train_test_split(
        X_test_full,
        y_test_full,
        train_size=TEST_SAMPLES,
        stratify=y_test_full,
        random_state=42
    )

else:

    X_test = X_test_full
    y_test = y_test_full


print("\nDataset prepared successfully.")

print("Training samples   :", len(X_train))
print("Validation samples :", len(X_val))
print("Testing samples    :", len(X_test))

print("Image shape        :", X_train.shape[1:])


# =========================
# 9. DISPLAY SAMPLE IMAGES
# =========================

def show_sample_images():

    plt.figure(figsize=(12, 5))

    for i in range(10):

        plt.subplot(2, 5, i + 1)

        plt.imshow(X_train[i], cmap="gray")

        plt.title(CLASS_NAMES[y_train[i]])

        plt.axis("off")

    plt.tight_layout()
    plt.show()


# =========================
# 10. BUILD FEEDFORWARD NN
# =========================

def build_model(
    hidden_units=(128, 64),
    activation="relu",
    learning_rate=0.01,
    optimizer_name="sgd"
):

    model = Sequential()

    # Input + first hidden layer
    model.add(
        Flatten(input_shape=(28, 28))
    )

    # Hidden layers
    for units in hidden_units:

        model.add(
            Dense(
                units,
                activation=activation
            )
        )

    # Output layer
    model.add(
        Dense(
            10,
            activation="softmax"
        )
    )

    # Optimizer
    if optimizer_name.lower() == "adam":

        optimizer = Adam(
            learning_rate=learning_rate
        )

    else:

        optimizer = SGD(
            learning_rate=learning_rate
        )

    # Compile
    model.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ============================================================
# 11. OPTION 1
# BASELINE FEEDFORWARD NEURAL NETWORK
# ============================================================

baseline_model = None
baseline_history = None
baseline_result = None


def train_baseline():

    global baseline_model
    global baseline_history
    global baseline_result

    print("\n" + "=" * 70)
    print("BASELINE FEEDFORWARD NEURAL NETWORK")
    print("=" * 70)

    baseline_model = build_model(
        hidden_units=(128, 64),
        activation="relu",
        learning_rate=0.01,
        optimizer_name="sgd"
    )

    print("\nNetwork Architecture:")
    baseline_model.summary()

    print("\nTraining model...")

    baseline_history = baseline_model.fit(
        X_train,
        y_train,
        validation_data=(X_val, y_val),
        epochs=BASELINE_EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=1
    )

    test_loss, test_accuracy = baseline_model.evaluate(
        X_test,
        y_test,
        verbose=0
    )

    baseline_result = {
        "loss": test_loss,
        "accuracy": test_accuracy
    }

    print("\n" + "-" * 50)
    print("BASELINE RESULTS")
    print("-" * 50)

    print(f"Test Loss     : {test_loss:.4f}")
    print(f"Test Accuracy : {test_accuracy * 100:.2f}%")

    return baseline_model


# ============================================================
# 12. LEARNING RATE DIAGNOSTIC
# ============================================================

learning_rate_results = {}


def classify_learning_rate(history):

    losses = np.array(history.history["loss"])

    if len(losses) == 0:
        return "Unknown"

    if not np.all(np.isfinite(losses)):
        return "FAILURE - Unstable / Too Large"

    initial_loss = losses[0]
    final_loss = losses[-1]

    max_loss = np.max(losses)

    improvement = (
        (initial_loss - final_loss)
        / max(initial_loss, 1e-8)
    )

    # Detect unstable behavior
    if max_loss > initial_loss * 2:

        return "FAILURE - Too Large"

    # Detect repeated increasing loss
    if len(losses) >= 3:

        last_three = losses[-3:]

        if (
            last_three[1] > last_three[0]
            and last_three[2] > last_three[1]
        ):

            return "WARNING - Unstable"

    # Very small learning rate
    if improvement < 0.05:

        return "WARNING - Too Small / Slow"

    return "STABLE"


def run_learning_rate_test():

    global learning_rate_results

    print("\n" + "=" * 70)
    print("LEARNING RATE FAILURE DETECTOR")
    print("=" * 70)

    learning_rates = [
        0.0001,
        0.001,
        0.01,
        0.1
    ]

    results = []

    for lr in learning_rates:

        print("\nTesting learning rate:", lr)

        model = build_model(
            hidden_units=(128, 64),
            activation="relu",
            learning_rate=lr,
            optimizer_name="sgd"
        )

        history = model.fit(
            X_train,
            y_train,
            validation_data=(X_val, y_val),
            epochs=LR_EPOCHS,
            batch_size=BATCH_SIZE,
            verbose=0
        )

        loss_history = history.history["loss"]
        accuracy_history = history.history["accuracy"]

        final_loss = loss_history[-1]
        final_accuracy = accuracy_history[-1]

        status = classify_learning_rate(history)

        results.append({
            "Learning Rate": lr,
            "Final Loss": final_loss,
            "Final Accuracy": final_accuracy * 100,
            "Status": status
        })

        learning_rate_results[lr] = {
            "model": model,
            "history": history
        }

        print("Final Loss     :", round(final_loss, 4))
        print("Final Accuracy :", round(final_accuracy * 100, 2), "%")
        print("Diagnosis      :", status)

    df = pd.DataFrame(results)

    print("\n" + "-" * 70)
    print("LEARNING RATE DIAGNOSTIC TABLE")
    print("-" * 70)

    print(df.to_string(index=False))

    # Plot
    plt.figure(figsize=(9, 5))

    for lr in learning_rates:

        history = learning_rate_results[lr]["history"]

        plt.plot(
            history.history["loss"],
            label=f"LR={lr}"
        )

    plt.xlabel("Epoch")
    plt.ylabel("Training Loss")
    plt.title("Learning Rate Comparison")
    plt.legend()
    plt.grid(True)
    plt.show()

    return df


# ============================================================
# 13. GRADIENT MONITOR
# ============================================================

class GradientMonitor(tf.keras.callbacks.Callback):

    def __init__(self, x_batch, y_batch):

        super().__init__()

        self.x_batch = tf.convert_to_tensor(
            x_batch,
            dtype=tf.float32
        )

        self.y_batch = tf.convert_to_tensor(
            y_batch,
            dtype=tf.int32
        )

        self.gradient_norms = []
        self.max_gradients = []

        self.loss_function = SparseCategoricalCrossentropy()

    def on_epoch_end(self, epoch, logs=None):

        with tf.GradientTape() as tape:

            predictions = self.model(
                self.x_batch,
                training=True
            )

            loss = self.loss_function(
                self.y_batch,
                predictions
            )

        gradients = tape.gradient(
            loss,
            self.model.trainable_variables
        )

        valid_gradients = [
            g for g in gradients
            if g is not None
        ]

        if len(valid_gradients) == 0:

            global_norm = 0.0
            max_gradient = 0.0

        else:

            global_norm = float(
                tf.linalg.global_norm(
                    valid_gradients
                ).numpy()
            )

            max_gradient = max(
                float(
                    tf.reduce_max(
                        tf.abs(g)
                    ).numpy()
                )
                for g in valid_gradients
            )

        self.gradient_norms.append(
            global_norm
        )

        self.max_gradients.append(
            max_gradient
        )


def gradient_status(norm):

    if not np.isfinite(norm):

        return "FAILURE - Exploding / NaN"

    if norm < 1e-5:

        return "WARNING - Vanishing Gradient"

    if norm > 10:

        return "WARNING - Exploding Gradient"

    return "STABLE"


# ============================================================
# 14. GRADIENT FAILURE DETECTOR
# ============================================================

gradient_results = {}


def run_gradient_test():

    global gradient_results

    print("\n" + "=" * 70)
    print("GRADIENT FAILURE DETECTOR")
    print("=" * 70)

    # Take a smaller batch for gradient inspection
    gradient_x = X_train[:512]
    gradient_y = y_train[:512]

    configurations = [

        {
            "name": "Normal ReLU Network",
            "hidden_units": (128, 64),
            "activation": "relu",
            "learning_rate": 0.01
        },

        {
            "name": "Deep Sigmoid Network",
            "hidden_units": (128, 128, 128, 128, 128),
            "activation": "sigmoid",
            "learning_rate": 0.01
        },

        {
            "name": "High Learning Rate Network",
            "hidden_units": (128, 64),
            "activation": "relu",
            "learning_rate": 0.5
        }
    ]

    summary = []

    for config in configurations:

        print("\nTesting:", config["name"])

        model = build_model(
            hidden_units=config["hidden_units"],
            activation=config["activation"],
            learning_rate=config["learning_rate"],
            optimizer_name="sgd"
        )

        monitor = GradientMonitor(
            gradient_x,
            gradient_y
        )

        history = model.fit(
            X_train[:5000],
            y_train[:5000],
            validation_data=(X_val[:1000], y_val[:1000]),
            epochs=GRADIENT_EPOCHS,
            batch_size=BATCH_SIZE,
            callbacks=[monitor],
            verbose=0
        )

        final_gradient = monitor.gradient_norms[-1]

        minimum_gradient = min(
            monitor.gradient_norms
        )

        maximum_gradient = max(
            monitor.gradient_norms
        )

        status = gradient_status(
            final_gradient
        )

        print(
            "Gradient Norms:",
            [round(x, 6) for x in monitor.gradient_norms]
        )

        print(
            "Final Gradient Norm:",
            round(final_gradient, 6)
        )

        print(
            "Diagnosis:",
            status
        )

        gradient_results[config["name"]] = {
            "model": model,
            "history": history,
            "monitor": monitor
        }

        summary.append({
            "Configuration": config["name"],
            "Minimum Gradient": minimum_gradient,
            "Maximum Gradient": maximum_gradient,
            "Final Gradient": final_gradient,
            "Status": status
        })

    df = pd.DataFrame(summary)

    print("\n" + "-" * 70)
    print("GRADIENT DIAGNOSTIC TABLE")
    print("-" * 70)

    print(df.to_string(index=False))

    # Plot each configuration separately
    for config in configurations:

        name = config["name"]

        monitor = gradient_results[name]["monitor"]

        plt.figure(figsize=(8, 5))

        plt.plot(
            range(1, len(monitor.gradient_norms) + 1),
            monitor.gradient_norms,
            marker="o"
        )

        plt.xlabel("Epoch")
        plt.ylabel("Gradient Norm")
        plt.title(
            "Gradient Norm - " + name
        )

        plt.grid(True)

        plt.show()

    return df


# ============================================================
# 15. OVERFITTING DETECTOR
# ============================================================

overfit_model = None
overfit_history = None


def detect_overfitting(history):

    train_loss = np.array(
        history.history["loss"]
    )

    val_loss = np.array(
        history.history["val_loss"]
    )

    train_accuracy = np.array(
        history.history["accuracy"]
    )

    val_accuracy = np.array(
        history.history["val_accuracy"]
    )

    best_val_loss_index = np.argmin(
        val_loss
    )

    best_val_loss = val_loss[
        best_val_loss_index
    ]

    final_train_loss = train_loss[-1]
    final_val_loss = val_loss[-1]

    final_train_accuracy = train_accuracy[-1]
    final_val_accuracy = val_accuracy[-1]

    accuracy_gap = (
        final_train_accuracy
        - final_val_accuracy
    )

    validation_loss_increase = (
        final_val_loss
        - best_val_loss
    )

    # Determine overfitting
    if (
        accuracy_gap > 0.08
        or validation_loss_increase > 0.10
    ):

        diagnosis = "FAILURE - Possible Overfitting"

    elif (
        accuracy_gap > 0.05
        or validation_loss_increase > 0.05
    ):

        diagnosis = "WARNING - Mild Overfitting"

    else:

        diagnosis = "STABLE - No Strong Overfitting"

    return {
        "best_epoch": best_val_loss_index + 1,
        "best_val_loss": best_val_loss,
        "final_train_loss": final_train_loss,
        "final_val_loss": final_val_loss,
        "final_train_accuracy": final_train_accuracy,
        "final_val_accuracy": final_val_accuracy,
        "accuracy_gap": accuracy_gap,
        "validation_loss_increase": validation_loss_increase,
        "diagnosis": diagnosis
    }


def run_overfitting_test():

    global overfit_model
    global overfit_history

    print("\n" + "=" * 70)
    print("OVERFITTING FAILURE DETECTOR")
    print("=" * 70)

    print("\nTraining a larger network for more epochs...")

    overfit_model = build_model(
        hidden_units=(256, 128, 64),
        activation="relu",
        learning_rate=0.01,
        optimizer_name="sgd"
    )

    overfit_history = overfit_model.fit(
        X_train,
        y_train,
        validation_data=(X_val, y_val),
        epochs=OVERFIT_EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=1
    )

    diagnosis = detect_overfitting(
        overfit_history
    )

    print("\n" + "-" * 70)
    print("OVERFITTING ANALYSIS")
    print("-" * 70)

    print(
        "Best Validation Epoch     :",
        diagnosis["best_epoch"]
    )

    print(
        "Best Validation Loss      :",
        round(
            diagnosis["best_val_loss"],
            4
        )
    )

    print(
        "Final Training Loss       :",
        round(
            diagnosis["final_train_loss"],
            4
        )
    )

    print(
        "Final Validation Loss     :",
        round(
            diagnosis["final_val_loss"],
            4
        )
    )

    print(
        "Final Training Accuracy   :",
        round(
            diagnosis["final_train_accuracy"] * 100,
            2
        ),
        "%"
    )

    print(
        "Final Validation Accuracy :",
        round(
            diagnosis["final_val_accuracy"] * 100,
            2
        ),
        "%"
    )

    print(
        "Train-Validation Gap      :",
        round(
            diagnosis["accuracy_gap"] * 100,
            2
        ),
        "%"
    )

    print(
        "Validation Loss Increase  :",
        round(
            diagnosis["validation_loss_increase"],
            4
        )
    )

    print(
        "\nDIAGNOSIS:",
        diagnosis["diagnosis"]
    )

    # Training loss plot
    plt.figure(figsize=(9, 5))

    plt.plot(
        overfit_history.history["loss"],
        label="Training Loss"
    )

    plt.plot(
        overfit_history.history["val_loss"],
        label="Validation Loss"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Loss")

    plt.title(
        "Training vs Validation Loss"
    )

    plt.legend()
    plt.grid(True)

    plt.show()

    # Accuracy plot
    plt.figure(figsize=(9, 5))

    plt.plot(
        overfit_history.history["accuracy"],
        label="Training Accuracy"
    )

    plt.plot(
        overfit_history.history["val_accuracy"],
        label="Validation Accuracy"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")

    plt.title(
        "Training vs Validation Accuracy"
    )

    plt.legend()
    plt.grid(True)

    plt.show()

    return diagnosis


# ============================================================
# 16. TRAINING CURVE ANALYZER
# ============================================================

def training_curve_analyzer():

    print("\n" + "=" * 70)
    print("TRAINING CURVE ANALYZER")
    print("=" * 70)

    global baseline_history

    if baseline_history is None:

        print(
            "\nBaseline model has not been trained."
        )

        print(
            "Training baseline model first..."
        )

        train_baseline()

    history = baseline_history.history

    # Loss curve
    plt.figure(figsize=(9, 5))

    plt.plot(
        history["loss"],
        label="Training Loss"
    )

    plt.plot(
        history["val_loss"],
        label="Validation Loss"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Loss")

    plt.title(
        "Baseline Training and Validation Loss"
    )

    plt.legend()
    plt.grid(True)

    plt.show()

    # Accuracy curve
    plt.figure(figsize=(9, 5))

    plt.plot(
        history["accuracy"],
        label="Training Accuracy"
    )

    plt.plot(
        history["val_accuracy"],
        label="Validation Accuracy"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")

    plt.title(
        "Baseline Training and Validation Accuracy"
    )

    plt.legend()
    plt.grid(True)

    plt.show()

    # Analyze convergence
    final_train_loss = history["loss"][-1]
    final_val_loss = history["val_loss"][-1]

    final_train_accuracy = history["accuracy"][-1]
    final_val_accuracy = history["val_accuracy"][-1]

    print("\nTraining Curve Analysis:")

    print(
        f"Final Training Loss   : {final_train_loss:.4f}"
    )

    print(
        f"Final Validation Loss : {final_val_loss:.4f}"
    )

    print(
        f"Final Training Accuracy   : "
        f"{final_train_accuracy * 100:.2f}%"
    )

    print(
        f"Final Validation Accuracy : "
        f"{final_val_accuracy * 100:.2f}%"
    )


# ============================================================
# 17. COMPLETE DIAGNOSTIC REPORT
# ============================================================

def complete_diagnostic_report():

    print("\n")
    print("=" * 70)
    print("COMPLETE FNN DIAGNOSTIC REPORT")
    print("=" * 70)

    print("\nRunning all major diagnostic tests...")

    # Baseline
    if baseline_model is None:

        print("\n[1/3] Running baseline test...")

        train_baseline()

    else:

        print("\n[1/3] Baseline already available.")

    # Learning rate
    if len(learning_rate_results) == 0:

        print("\n[2/3] Running learning rate test...")

        lr_df = run_learning_rate_test()

    else:

        print("\n[2/3] Learning rate results already available.")

        lr_rows = []

        for lr, data in learning_rate_results.items():

            history = data["history"]

            lr_rows.append({
                "Learning Rate": lr,
                "Final Loss": history.history["loss"][-1],
                "Final Accuracy":
                    history.history["accuracy"][-1] * 100,
                "Status":
                    classify_learning_rate(history)
            })

        lr_df = pd.DataFrame(lr_rows)

    # Overfitting
    print("\n[3/3] Running overfitting test...")

    overfit_diagnosis = run_overfitting_test()

    # Final summary
    print("\n")
    print("=" * 70)
    print("FINAL DIAGNOSTIC SUMMARY")
    print("=" * 70)

    print("\n1. BASELINE MODEL")
    print("-" * 40)

    if baseline_result is not None:

        print(
            "Test Accuracy:",
            round(
                baseline_result["accuracy"] * 100,
                2
            ),
            "%"
        )

    print("\n2. LEARNING RATE DIAGNOSIS")
    print("-" * 40)

    print(
        lr_df[
            [
                "Learning Rate",
                "Final Accuracy",
                "Status"
            ]
        ].to_string(index=False)
    )

    print("\n3. OVERFITTING DIAGNOSIS")
    print("-" * 40)

    print(
        overfit_diagnosis["diagnosis"]
    )

    print("\n4. RECOMMENDATIONS")
    print("-" * 40)

    print("""
• If learning rate is too small:
  Increase the learning rate gradually.

• If learning rate is too large:
  Reduce the learning rate.

• If gradients are vanishing:
  Consider ReLU, better initialization,
  normalization or a shallower/deeper architecture.

• If gradients are exploding:
  Reduce learning rate, use gradient clipping
  or improve network initialization.

• If overfitting occurs:
  Use dropout, regularization, early stopping,
  data augmentation or a smaller network.

• Always compare training and validation curves
  instead of looking only at training accuracy.
""")


# ============================================================
# 18. PREDICTION DEMO
# ============================================================

def prediction_demo():

    global baseline_model

    if baseline_model is None:

        print(
            "\nBaseline model is not trained."
        )

        print(
            "Training baseline model..."
        )

        train_baseline()

    predictions = baseline_model.predict(
        X_test[:10],
        verbose=0
    )

    predicted_classes = np.argmax(
        predictions,
        axis=1
    )

    print("\n" + "=" * 70)
    print("PREDICTION DEMO")
    print("=" * 70)

    for i in range(10):

        actual = CLASS_NAMES[
            y_test[i]
        ]

        predicted = CLASS_NAMES[
            predicted_classes[i]
        ]

        confidence = (
            predictions[i][predicted_classes[i]]
            * 100
        )

        print(
            f"\nImage {i + 1}"
        )

        print(
            "Actual    :",
            actual
        )

        print(
            "Predicted :",
            predicted
        )

        print(
            "Confidence:",
            f"{confidence:.2f}%"
        )


# ============================================================
# 19. MAIN MENU
# ============================================================

def main_menu():

    while True:

        print("\n")
        print("=" * 70)
        print("FASHION-MNIST FNN DIAGNOSTIC SYSTEM")
        print("=" * 70)

        print("""
1. Train Baseline Feedforward Neural Network
2. Learning Rate Failure Detector
3. Gradient Failure Detector
4. Overfitting Detector
5. Training Curve Analyzer
6. Complete Diagnostic Report
7. Prediction Demo
8. Show Fashion-MNIST Sample Images
9. Exit
""")

        choice = input(
            "Enter your choice (1-9): "
        )

        if choice == "1":

            train_baseline()

        elif choice == "2":

            run_learning_rate_test()

        elif choice == "3":

            run_gradient_test()

        elif choice == "4":

            run_overfitting_test()

        elif choice == "5":

            training_curve_analyzer()

        elif choice == "6":

            complete_diagnostic_report()

        elif choice == "7":

            prediction_demo()

        elif choice == "8":

            show_sample_images()

        elif choice == "9":

            print(
                "\nExiting the diagnostic system."
            )

            print(
                "Exercise 3 completed successfully."
            )

            break

        else:

            print(
                "\nInvalid choice."
            )

            print(
                "Please enter a number from 1 to 9."
            )


# ============================================================
# 20. START PROGRAM
# ============================================================

print("\n")
print("=" * 70)
print("EXERCISE 3 READY")
print("=" * 70)

print("""
Feedforward Neural Network Diagnostic System

Dataset   : Fashion-MNIST
Input     : 28 x 28 = 784 pixels
Output    : 10 classes
Optimizer : SGD / Adam
Diagnostics:
    - Learning Rate
    - Gradient
    - Overfitting
    - Training Curves

The program is ready.
""")

main_menu()
```
