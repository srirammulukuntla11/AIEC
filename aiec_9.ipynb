```python
# ================================================================
# Exercise 9: Formulate a Hopfield Neural Network Associative Memory
# and Pattern Reconstruction System using the Optical Recognition
# of Handwritten Digits Dataset
# ================================================================

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import random
import warnings
from sklearn.datasets import load_digits

warnings.filterwarnings("ignore")

# Reproducibility
np.random.seed(42)
random.seed(42)

# ================================================================
# 1. LOAD DATASET
# ================================================================

print("=" * 85)
print("EXERCISE 9: HOPFIELD NEURAL NETWORK ASSOCIATIVE MEMORY")
print("AND HANDWRITTEN DIGIT PATTERN RECONSTRUCTION SYSTEM")
print("=" * 85)

digits_data = load_digits()

images = digits_data.images
labels = digits_data.target

print("\nDataset loaded successfully!")
print("Dataset: Optical Recognition of Handwritten Digits")
print("Total images:", len(images))
print("Image dimensions:", images.shape[1], "x", images.shape[2])
print("Available digits:", np.unique(labels))
print("Pixel intensity range: 0 to 16")

# Convert grayscale images to bipolar patterns.
# Active pixels = +1; inactive pixels = -1.
# Threshold 8 separates darker from brighter pixels.

binary_images = np.where(images > 8, 1, -1)
binary_patterns = binary_images.reshape(len(images), -1)

# ================================================================
# 2. SELECT REPRESENTATIVE PATTERN FOR EACH DIGIT
# ================================================================

def select_representative_patterns():
    """
    Select one representative 8x8 pattern for each digit.
    The selected pattern is the sample closest to the
    majority bipolar pattern for that digit.
    """

    selected = {}

    for digit in range(10):
        digit_patterns = binary_patterns[labels == digit]

        majority_pattern = np.where(
            np.sum(digit_patterns, axis=0) >= 0, 1, -1
        )

        distances = np.sum(
            digit_patterns != majority_pattern, axis=1
        )

        selected_index = np.argmin(distances)
        selected[digit] = digit_patterns[selected_index].copy()

    return selected


stored_patterns = {}
weight_matrix = None

# ================================================================
# 3. HOPFIELD NEURAL NETWORK
# ================================================================

class HopfieldNetwork:

    def __init__(self, patterns):
        self.patterns = np.asarray(patterns, dtype=int)
        self.num_neurons = self.patterns.shape[1]
        self.weights = np.zeros(
            (self.num_neurons, self.num_neurons),
            dtype=float
        )
        self.train()

    def train(self):
        """
        Hebbian learning rule:
        W = (1/N) * sum(x_p * x_p^T)
        Self-connections are removed by setting the diagonal to zero.
        """

        self.weights = (
            self.patterns.T @ self.patterns
        ) / len(self.patterns)

        np.fill_diagonal(self.weights, 0)

    def energy(self, pattern):
        """Calculate Hopfield network energy."""

        pattern = np.asarray(pattern, dtype=float)

        return -0.5 * pattern @ self.weights @ pattern

    def recall(self, input_pattern, max_iterations=100):
        """
        Asynchronous neuron updates.
        Zero-valued missing pixels are allowed as input;
        each updated neuron becomes either +1 or -1.
        """

        state = np.asarray(input_pattern, dtype=int).copy()
        state = state.flatten()

        energies = [self.energy(state)]

        for iteration in range(1, max_iterations + 1):

            previous_state = state.copy()

            # Update neurons in a random order.
            for neuron in np.random.permutation(self.num_neurons):

                activation = np.dot(
                    self.weights[neuron], state
                )

                if activation > 0:
                    state[neuron] = 1
                elif activation < 0:
                    state[neuron] = -1
                # If activation is zero, keep the existing value.

            energies.append(self.energy(state))

            if np.array_equal(state, previous_state):
                return state, iteration, energies

        return state, max_iterations, energies


# ================================================================
# 4. HELPER FUNCTIONS
# ================================================================

def display_patterns(pattern_list, titles, rows=1, cols=None):
    """Display one or more bipolar patterns."""

    if cols is None:
        cols = len(pattern_list)

    rows = max(rows, int(np.ceil(len(pattern_list) / cols)))

    fig, axes = plt.subplots(
        rows, cols,
        figsize=(3 * cols, 3 * rows),
        squeeze=False
    )

    for i, ax in enumerate(axes.flat):

        if i < len(pattern_list):
            pattern = np.asarray(pattern_list[i]).reshape(8, 8)

            ax.imshow(
                pattern,
                cmap="gray",
                vmin=-1,
                vmax=1,
                interpolation="nearest"
            )

            ax.set_title(titles[i])
            ax.set_xticks(range(8))
            ax.set_yticks(range(8))
            ax.set_xticklabels([])
            ax.set_yticklabels([])
            ax.grid(True, linewidth=0.3)

        else:
            ax.axis("off")

    plt.tight_layout()
    plt.show()


def display_weight_matrix(weights):
    plt.figure(figsize=(8, 6))
    plt.imshow(weights, cmap="coolwarm", aspect="auto")
    plt.colorbar(label="Connection Weight")
    plt.title("Hopfield Network Weight Matrix")
    plt.xlabel("Neuron")
    plt.ylabel("Neuron")
    plt.tight_layout()
    plt.show()


def create_network(pattern_dictionary):
    if not pattern_dictionary:
        raise ValueError("No patterns are stored. Choose menu option 1 first.")

    digits = sorted(pattern_dictionary.keys())
    patterns = np.array(
        [pattern_dictionary[d] for d in digits]
    )

    return HopfieldNetwork(patterns), digits


def corrupt_pattern(pattern, percentage):
    """Randomly flip a percentage of bipolar pixels."""

    result = np.asarray(pattern, dtype=int).copy()
    number_to_flip = int(round(len(result) * percentage / 100))

    if number_to_flip > 0:
        positions = np.random.choice(
            len(result),
            size=number_to_flip,
            replace=False
        )
        result[positions] *= -1

    return result, number_to_flip


def identify_digit(recalled_pattern, pattern_dictionary):
    """Identify an exact stored pattern, if present."""

    for digit, pattern in pattern_dictionary.items():
        if np.array_equal(recalled_pattern, pattern):
            return digit

    return None


def pixel_accuracy(original, reconstructed):
    return 100 * np.mean(
        np.asarray(original) == np.asarray(reconstructed)
    )


def get_digit_choice(pattern_dictionary):
    available = sorted(pattern_dictionary.keys())

    print("\nStored digits:", available)

    while True:
        try:
            digit = int(input("Enter the digit to use: "))

            if digit in pattern_dictionary:
                return digit

            print("That digit is not stored. Choose from", available)

        except ValueError:
            print("Please enter an integer digit.")


def ensure_patterns_stored():
    global stored_patterns, weight_matrix

    if not stored_patterns:
        print("\nNo patterns are stored yet.")
        print("Automatically storing representative patterns for digits 0-9...")

        stored_patterns = select_representative_patterns()
        network, _ = create_network(stored_patterns)
        weight_matrix = network.weights.copy()

        print("Stored 10 representative digit patterns.")


# ================================================================
# MENU OPTION 1: STORE HANDWRITTEN DIGIT PATTERNS
# ================================================================

def option_1():
    global stored_patterns, weight_matrix

    print("\n" + "=" * 65)
    print("1. STORE HANDWRITTEN DIGIT PATTERNS")
    print("=" * 65)

    stored_patterns = select_representative_patterns()

    network, digits = create_network(stored_patterns)
    weight_matrix = network.weights.copy()

    print("\nHebbian learning completed successfully!")
    print("Number of stored patterns:", len(stored_patterns))
    print("Neurons in the network:", network.num_neurons)
    print("Weight matrix shape:", network.weights.shape)

    print("\nStored bipolar patterns:")
    for digit in digits:
        print(f"\nDigit {digit}:")
        print(stored_patterns[digit].reshape(8, 8))

    display_patterns(
        [stored_patterns[d] for d in digits],
        [f"Stored Digit {d}" for d in digits],
        rows=2,
        cols=5
    )

    display_weight_matrix(network.weights)

    print("Diagonal self-connections are zero.")
    print("Weight matrix is symmetric:",
          np.allclose(network.weights, network.weights.T))


# ================================================================
# MENU OPTION 2: RECALL A NOISY DIGIT PATTERN
# ================================================================

def option_2():
    ensure_patterns_stored()

    print("\n" + "=" * 65)
    print("2. RECALL A NOISY DIGIT PATTERN")
    print("=" * 65)

    digit = get_digit_choice(stored_patterns)

    print("\nChoose corruption level:")
    print("1. 5%")
    print("2. 10%")
    print("3. 20%")
    print("4. 30%")

    levels = {1: 5, 2: 10, 3: 20, 4: 30}

    try:
        choice = int(input("Enter choice (1-4): "))
        if choice not in levels:
            print("Invalid choice. Using 10% corruption.")
            percentage = 10
        else:
            percentage = levels[choice]
    except ValueError:
        percentage = 10

    original = stored_patterns[digit]
    noisy, flipped = corrupt_pattern(original, percentage)

    network, _ = create_network(stored_patterns)
    recalled, iterations, energies = network.recall(noisy)

    matched_digit = identify_digit(recalled, stored_patterns)
    accuracy = pixel_accuracy(original, recalled)

    print("\nOriginal digit:", digit)
    print("Corruption percentage:", percentage, "%")
    print("Number of flipped pixels:", flipped)
    print("Recall iterations:", iterations)
    print("Pixel reconstruction accuracy:", round(accuracy, 2), "%")
    print("Recalled digit:",
          matched_digit if matched_digit is not None else "No exact stored match")
    print("Exact recall successful:", np.array_equal(original, recalled))

    display_patterns(
        [original, noisy, recalled],
        [f"Original Digit {digit}",
         f"Noisy Input ({percentage}%)",
         "Recalled Pattern"]
    )

    plt.figure(figsize=(7, 4))
    plt.plot(range(len(energies)), energies, marker="o")
    plt.title("Hopfield Energy During Recall")
    plt.xlabel("Update Cycle")
    plt.ylabel("Network Energy")
    plt.grid(True)
    plt.tight_layout()
    plt.show()


# ================================================================
# MENU OPTION 3: MISSING PIXEL RECONSTRUCTION CHALLENGE
# ================================================================

def option_3():
    ensure_patterns_stored()

    print("\n" + "=" * 65)
    print("3. MISSING PIXEL RECONSTRUCTION CHALLENGE")
    print("=" * 65)

    digit = get_digit_choice(stored_patterns)
    original = stored_patterns[digit].copy().reshape(8, 8)

    print("\nChoose the missing region:")
    print("1. One row")
    print("2. One column")
    print("3. Rectangular block")

    try:
        choice = int(input("Enter choice (1-3): "))
    except ValueError:
        choice = 1

    incomplete = original.copy()
    mask = np.zeros((8, 8), dtype=bool)

    if choice == 1:
        try:
            row = int(input("Enter row to remove (0-7): "))
            row = int(np.clip(row, 0, 7))
        except ValueError:
            row = 3

        mask[row, :] = True

    elif choice == 2:
        try:
            col = int(input("Enter column to remove (0-7): "))
            col = int(np.clip(col, 0, 7))
        except ValueError:
            col = 3

        mask[:, col] = True

    else:
        try:
            row = int(input("Enter starting row (0-7): "))
            col = int(input("Enter starting column (0-7): "))
            height = int(input("Enter block height (1-8): "))
            width = int(input("Enter block width (1-8): "))

            row = int(np.clip(row, 0, 7))
            col = int(np.clip(col, 0, 7))
            height = int(np.clip(height, 1, 8))
            width = int(np.clip(width, 1, 8))

        except ValueError:
            row, col, height, width = 2, 2, 3, 3

        mask[
            row:min(8, row + height),
            col:min(8, col + width)
        ] = True

    # Zero means a missing/unknown pixel.
    incomplete[mask] = 0

    network, _ = create_network(stored_patterns)
    reconstructed, iterations, _ = network.recall(
        incomplete.flatten()
    )

    reconstructed_image = reconstructed.reshape(8, 8)

    full_accuracy = pixel_accuracy(
        original.flatten(), reconstructed
    )

    missing_accuracy = (
        100 * np.mean(
            original[mask] == reconstructed_image[mask]
        )
        if np.any(mask) else 100
    )

    print("\nMissing pixels:", int(np.sum(mask)))
    print("Recall iterations:", iterations)
    print("Full-image accuracy:", round(full_accuracy, 2), "%")
    print("Reconstructed missing-region accuracy:",
          round(missing_accuracy, 2), "%")

    display_patterns(
        [original.flatten(), incomplete.flatten(), reconstructed],
        ["Original Digit", "Incomplete Pattern", "Reconstructed Digit"]
    )

    # Search for the largest missing rectangle that achieves at
    # least 80% correct pixels within the missing region.
    # This is an empirical search over rectangles and locations.
    best_area = 0
    best_details = None

    for height in range(1, 9):
        for width in range(1, 9):
            area = height * width

            if area <= best_area:
                continue

            for r in range(9 - height):
                for c in range(9 - width):

                    test_input = original.copy()
                    region = np.zeros((8, 8), dtype=bool)

                    region[r:r + height, c:c + width] = True
                    test_input[region] = 0

                    test_recalled, _, _ = network.recall(
                        test_input.flatten(),
                        max_iterations=100
                    )

                    test_image = test_recalled.reshape(8, 8)

                    region_accuracy = 100 * np.mean(
                        original[region] == test_image[region]
                    )

                    if region_accuracy >= 80:
                        best_area = area
                        best_details = (
                            height, width, r, c, region_accuracy
                        )

    print("\nLargest successfully reconstructed rectangle")
    print("Success criterion: at least 80% correct pixels in the missing region.")

    if best_details:
        h, w, r, c, acc = best_details
        print(f"Rectangle size: {h} x {w}")
        print(f"Starting location: row {r}, column {c}")
        print("Missing-region accuracy:", round(acc, 2), "%")
        print("Area:", best_area, "pixels")
    else:
        print("No rectangle met the reconstruction criterion.")


# ================================================================
# MENU OPTION 4: DIGIT CORRUPTION TOLERANCE TEST
# ================================================================

def option_4():
    ensure_patterns_stored()

    print("\n" + "=" * 65)
    print("4. DIGIT CORRUPTION TOLERANCE TEST")
    print("=" * 65)

    levels = [5, 10, 20, 30, 40, 50]
    network, digits = create_network(stored_patterns)

    results = []

    for digit in digits:
        original = stored_patterns[digit]
        maximum_success = 0

        print(f"\nTesting digit {digit}...")

        for percentage in levels:
            successes = 0
            trials = 20

            for _ in range(trials):
                noisy, _ = corrupt_pattern(original, percentage)
                recalled, _, _ = network.recall(noisy)

                if np.array_equal(recalled, original):
                    successes += 1

            accuracy = 100 * successes / trials

            if successes == trials:
                maximum_success = percentage

            results.append({
                "Digit": digit,
                "Corruption (%)": percentage,
                "Successful Recalls": successes,
                "Trials": trials,
                "Recall Accuracy (%)": accuracy
            })

        print("Highest tested level with 20/20 successful recalls:",
              f"{maximum_success}%")

    df = pd.DataFrame(results)
    print("\nDetailed corruption tolerance results:")
    print(df.to_string(index=False))

    summary = (
        df.groupby("Digit")
        .apply(
            lambda group: pd.Series({
                "Best Tested Corruption (%)": max(
                    [p for p in levels
                     if (
                         group.loc[
                             group["Corruption (%)"] == p,
                             "Successful Recalls"
                         ].iloc[0] == 20
                     )],
                    default=0
                ),
                "Mean Recall Accuracy (%)":
                    group["Recall Accuracy (%)"].mean()
            }),
            include_groups=False
        )
        .reset_index()
    )

    summary = summary.sort_values(
        ["Best Tested Corruption (%)", "Mean Recall Accuracy (%)"],
        ascending=False
    )

    print("\nRanking: most noise-tolerant to least noise-tolerant")
    print(summary.to_string(index=False))

    plt.figure(figsize=(9, 5))

    for digit in digits:
        subset = df[df["Digit"] == digit]
        plt.plot(
            subset["Corruption (%)"],
            subset["Recall Accuracy (%)"],
            marker="o",
            label=str(digit)
        )

    plt.title("Hopfield Recall Accuracy vs. Corruption Level")
    plt.xlabel("Corruption (%)")
    plt.ylabel("Exact Recall Accuracy (%)")
    plt.legend(title="Digit", ncol=2)
    plt.grid(True)
    plt.tight_layout()
    plt.show()


# ================================================================
# MENU OPTION 5: HOPFIELD MEMORY CAPACITY CHALLENGE
# ================================================================

def option_5():
    ensure_patterns_stored()

    print("\n" + "=" * 65)
    print("5. HOPFIELD MEMORY CAPACITY CHALLENGE")
    print("=" * 65)

    requested_counts = [2, 3, 4, 5, 6, 8, 10]
    all_digits = sorted(stored_patterns.keys())
    results = []

    for count in requested_counts:
        selected_digits = all_digits[:count]

        patterns = np.array([
            stored_patterns[d] for d in selected_digits
        ])

        network = HopfieldNetwork(patterns)

        successes = 0
        total = 0
        trials_per_digit = 10

        for digit in selected_digits:
            original = stored_patterns[digit]

            for _ in range(trials_per_digit):
                noisy, _ = corrupt_pattern(original, 10)
                recalled, _, _ = network.recall(noisy)

                total += 1

                if np.array_equal(recalled, original):
                    successes += 1

        failures = total - successes
        accuracy = 100 * successes / total

        results.append({
            "Stored Patterns": count,
            "Successful Recalls": successes,
            "Failed Recalls": failures,
            "Total Tests": total,
            "Recall Accuracy (%)": round(accuracy, 2)
        })

    df = pd.DataFrame(results)

    print("\nMemory capacity experiment:")
    print(df.to_string(index=False))

    best_row = df.loc[df["Recall Accuracy (%)"].idxmax()]
    print("\nBest tested configuration:",
          int(best_row["Stored Patterns"]), "patterns")
    print("Best recall accuracy:",
          best_row["Recall Accuracy (%)"], "%")

    reduced = False

    for i in range(1, len(df)):
        previous = df.iloc[i - 1]["Recall Accuracy (%)"]
        current = df.iloc[i]["Recall Accuracy (%)"]

        if current < previous:
            print(
                "Performance decreased when increasing the tested "
                f"pattern count from {int(df.iloc[i-1]['Stored Patterns'])} "
                f"to {int(df.iloc[i]['Stored Patterns'])}."
            )
            reduced = True

    if not reduced:
        print(
            "No decrease was observed between consecutive tested "
            "configurations. This does not mean capacity is unlimited."
        )

    plt.figure(figsize=(8, 5))
    plt.plot(
        df["Stored Patterns"],
        df["Recall Accuracy (%)"],
        marker="o"
    )
    plt.title("Hopfield Network Memory Capacity")
    plt.xlabel("Number of Stored Patterns")
    plt.ylabel("Noisy Recall Accuracy (%)")
    plt.xticks(requested_counts)
    plt.grid(True)
    plt.tight_layout()
    plt.show()


# ================================================================
# MENU OPTION 6: FALSE MEMORY DETECTOR
# ================================================================

def option_6():
    ensure_patterns_stored()

    print("\n" + "=" * 65)
    print("6. FALSE MEMORY DETECTOR")
    print("=" * 65)

    network, _ = create_network(stored_patterns)

    print("\nChoose input:")
    print("1. Corrupted stored digit")
    print("2. Random bipolar pattern")

    try:
        choice = int(input("Enter choice (1-2): "))
    except ValueError:
        choice = 2

    if choice == 1:
        digit = get_digit_choice(stored_patterns)
        original = stored_patterns[digit]

        noisy, flipped = corrupt_pattern(original, 30)

        print("\nOriginal digit:", digit)
        print("Pixels flipped:", flipped)

        input_pattern = noisy

    else:
        input_pattern = np.random.choice(
            [-1, 1], size=64
        )

        print("\nGenerated random bipolar pattern.")

    recalled, iterations, energies = network.recall(input_pattern)

    matched_digit = identify_digit(recalled, stored_patterns)

    if matched_digit is None:
        status = "SPURIOUS / FALSE MEMORY"
    else:
        status = f"Exact match with stored digit {matched_digit}"

    print("\nRecall iterations:", iterations)
    print("Final energy:", round(energies[-1], 2))
    print("Matching status:", status)

    print(
        "Stable state:",
        "Yes" if np.array_equal(recalled, network.recall(recalled)[0])
        else "Not confirmed"
    )

    display_patterns(
        [input_pattern, recalled],
        ["Input Pattern", "Final Stable Pattern"]
    )

    # Important: a pattern not matching a stored pattern is
    # classified as a spurious/false memory for this experiment.
    # A single recall trial cannot establish all possible attractors.


# ================================================================
# MENU OPTION 7: DIGIT CONFUSION ANALYZER
# ================================================================

def option_7():
    ensure_patterns_stored()

    print("\n" + "=" * 65)
    print("7. DIGIT CONFUSION ANALYZER")
    print("=" * 65)

    network, digits = create_network(stored_patterns)

    confusion = pd.DataFrame(
        0, index=digits, columns=digits, dtype=int
    )

    trials_per_digit = 30
    corruption_level = 20
    unknown_count = 0

    for digit in digits:
        original = stored_patterns[digit]

        for _ in range(trials_per_digit):
            noisy, _ = corrupt_pattern(
                original, corruption_level
            )

            recalled, _, _ = network.recall(noisy)
            matched_digit = identify_digit(
                recalled, stored_patterns
            )

            if matched_digit is not None:
                confusion.loc[digit, matched_digit] += 1
            else:
                unknown_count += 1

    print("\nConfusion matrix (rows = original, columns = recalled)")
    print(confusion)

    pairs = []

    for original_digit in digits:
        for recalled_digit in digits:
            if original_digit != recalled_digit:
                count = confusion.loc[
                    original_digit, recalled_digit
                ]

                if count > 0:
                    pairs.append((
                        original_digit,
                        recalled_digit,
                        int(count)
                    ))

    pairs.sort(key=lambda item: item[2], reverse=True)

    print("\nMost frequently confused digit pairs:")

    if pairs:
        for original_digit, recalled_digit, count in pairs[:10]:
            print(
                f"Digit {original_digit} -> Digit {recalled_digit}: "
                f"{count} times"
            )

        first = pairs[0]

        print(
            f"\nMost confused pair: Digit {first[0]} -> "
            f"Digit {first[1]} ({first[2]} times)"
        )

    else:
        print("No cross-digit confusion was observed.")

    print("Recalls not matching any stored pattern:", unknown_count)

    plt.figure(figsize=(8, 7))
    plt.imshow(confusion.values, cmap="Blues")
    plt.colorbar(label="Recall Count")
    plt.xticks(range(len(digits)), digits)
    plt.yticks(range(len(digits)), digits)
    plt.xlabel("Recalled Digit")
    plt.ylabel("Original Digit")
    plt.title(
        f"Digit Confusion Matrix ({corruption_level}% corruption)"
    )

    for i in range(len(digits)):
        for j in range(len(digits)):
            plt.text(
                j, i, str(confusion.iloc[i, j]),
                ha="center", va="center"
            )

    plt.tight_layout()
    plt.show()


# ================================================================
# MENU OPTION 8: HOPFIELD RECALL PERFORMANCE REPORT
# ================================================================

def option_8():
    ensure_patterns_stored()

    print("\n" + "=" * 65)
    print("8. HOPFIELD RECALL PERFORMANCE REPORT")
    print("=" * 65)

    network, digits = create_network(stored_patterns)

    percentages = [5, 10, 20, 30, 40, 50]
    report = []

    for digit in digits:
        original = stored_patterns[digit]

        for percentage in percentages:
            noisy, flipped = corrupt_pattern(
                original, percentage
            )

            recalled, iterations, _ = network.recall(noisy)

            matched_digit = identify_digit(
                recalled, stored_patterns
            )

            accuracy = pixel_accuracy(original, recalled)

            report.append({
                "Stored Digit": digit,
                "Corruption (%)": percentage,
                "Flipped Pixels": flipped,
                "Recall Iterations": iterations,
                "Pixel Accuracy (%)": round(accuracy, 2),
                "Recalled Digit": (
                    matched_digit
                    if matched_digit is not None
                    else "Unknown"
                ),
                "False Memory": matched_digit is None,
                "Exact Recall": np.array_equal(original, recalled)
            })

    df = pd.DataFrame(report)

    print("\nFull recall performance report:")
    print(df.to_string(index=False))

    print("\nSummary by corruption level:")

    summary = df.groupby("Corruption (%)").agg(
        Average_Pixel_Accuracy=("Pixel Accuracy (%)", "mean"),
        Exact_Recall_Rate=("Exact Recall", "mean"),
        False_Memory_Count=("False Memory", "sum"),
        Average_Iterations=("Recall Iterations", "mean")
    ).reset_index()

    summary["Exact Recall Rate (%)"] = (
        summary["Exact_Recall_Rate"] * 100
    )

    summary["Average_Pixel_Accuracy"] = (
        summary["Average_Pixel_Accuracy"].round(2)
    )

    summary["Average_Iterations"] = (
        summary["Average_Iterations"].round(2)
    )

    summary = summary.drop(columns=["Exact_Recall_Rate"])

    print(summary.to_string(index=False))

    plt.figure(figsize=(8, 5))
    plt.plot(
        summary["Corruption (%)"],
        summary["Average_Pixel_Accuracy"],
        marker="o"
    )
    plt.title("Average Pixel Reconstruction Accuracy")
    plt.xlabel("Corruption Level (%)")
    plt.ylabel("Average Pixel Accuracy (%)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 5))
    plt.plot(
        summary["Corruption (%)"],
        summary["False_Memory_Count"],
        marker="o"
    )
    plt.title("False Memories vs. Corruption Level")
    plt.xlabel("Corruption Level (%)")
    plt.ylabel("Number of False Memories")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    print("\nReport generated successfully.")


# ================================================================
# 5. MAIN MENU
# ================================================================

def main():
    menu = """
===============================================================
        HOPFIELD NEURAL NETWORK - EXERCISE 9
===============================================================
1. Store Handwritten Digit Patterns
2. Recall a Noisy Digit Pattern
3. Missing Pixel Reconstruction Challenge
4. Digit Corruption Tolerance Test
5. Hopfield Memory Capacity Challenge
6. False Memory Detector
7. Digit Confusion Analyzer
8. Hopfield Recall Performance Report
9. Exit
===============================================================
"""

    while True:
        print(menu)

        try:
            choice = int(input("Enter your choice (1-9): "))
        except ValueError:
            print("Invalid input. Enter a number from 1 to 9.")
            continue

        if choice == 1:
            option_1()

        elif choice == 2:
            option_2()

        elif choice == 3:
            option_3()

        elif choice == 4:
            option_4()

        elif choice == 5:
            option_5()

        elif choice == 6:
            option_6()

        elif choice == 7:
            option_7()

        elif choice == 8:
            option_8()

        elif choice == 9:
            print("\nExiting Hopfield Neural Network System.")
            print("Exercise 9 completed.")
            break

        else:
            print("Invalid choice. Please select 1-9.")


# ================================================================
# 6. RUN THE PROGRAM
# ================================================================

if __name__ == "__main__":
    main()
```
