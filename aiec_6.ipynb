```python
# ================================================================
# EXERCISE 6: BIDIRECTIONAL ASSOCIATIVE MEMORY (BAM)
# Handwritten Digit and Code Association System
# Dataset: Optical Recognition of Handwritten Digits
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split

# ================================================================
# 1. LOAD DATASET
# ================================================================

digits_data = load_digits()
images = digits_data.images
labels = digits_data.target

# Convert 8x8 images into bipolar vectors
X_all = np.where(images > 0, 1, -1).astype(int).reshape(-1, 64)
y_all = labels.astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X_all, y_all, test_size=0.25,
    random_state=42, stratify=y_all
)

# Select representative pattern for each digit
stored_patterns = {}

for digit in range(10):
    patterns = X_train[y_train == digit]
    mean_pattern = patterns.mean(axis=0)
    distances = np.sum((patterns - mean_pattern) ** 2, axis=1)
    stored_patterns[digit] = patterns[np.argmin(distances)].copy()

X_stored = np.array([stored_patterns[d] for d in range(10)])


# ================================================================
# 2. GENERATE UNIQUE BIPOLAR DIGIT CODES
# ================================================================

def digit_to_code(digit):
    code = -np.ones(10, dtype=int)
    code[int(digit)] = 1
    return code


def code_to_digit(code):
    code = np.asarray(code).flatten()
    positions = np.where(code == 1)[0]

    if len(positions) == 1 and np.all(np.isin(code, [-1, 1])):
        return int(positions[0])

    return None


Y_stored = np.array([digit_to_code(d) for d in range(10)])


# ================================================================
# 3. BAM NEURAL NETWORK
# ================================================================

class BAMNetwork:

    def __init__(self, X, Y, max_updates=30):
        self.X = np.asarray(X, dtype=int)
        self.Y = np.asarray(Y, dtype=int)
        self.max_updates = max_updates

        # Correlation learning rule
        self.W = self.X.T @ self.Y
        self.WT = self.W.T

    def step(self, values, previous=None):
        result = np.sign(values).astype(int)

        # Retain previous state for zero net input
        if previous is None:
            result[result == 0] = 1
        else:
            result[result == 0] = previous[result == 0]

        return result

    def forward_once(self, x):
        return self.step(np.asarray(x) @ self.W)

    def backward_once(self, y):
        return self.step(np.asarray(y) @ self.WT)

    def recall_forward(self, x, return_updates=False):
        x_state = np.asarray(x, dtype=int).flatten().copy()
        y_state = self.forward_once(x_state)
        updates = 1

        for _ in range(self.max_updates):
            new_x = self.step(y_state @ self.WT, x_state)
            new_y = self.step(new_x @ self.W, y_state)
            updates += 2

            if (np.array_equal(new_x, x_state)
                    and np.array_equal(new_y, y_state)):
                x_state, y_state = new_x, new_y
                break

            x_state, y_state = new_x, new_y

        if return_updates:
            return x_state, y_state, updates

        return x_state, y_state

    def recall_backward(self, y, return_updates=False):
        y_state = np.asarray(y, dtype=int).flatten().copy()
        x_state = self.backward_once(y_state)
        updates = 1

        for _ in range(self.max_updates):
            new_y = self.step(x_state @ self.W, y_state)
            new_x = self.step(new_y @ self.WT, x_state)
            updates += 2

            if (np.array_equal(new_y, y_state)
                    and np.array_equal(new_x, x_state)):
                y_state, x_state = new_y, new_x
                break

            y_state, x_state = new_y, new_x

        if return_updates:
            return y_state, x_state, updates

        return y_state, x_state


network = BAMNetwork(X_stored, Y_stored)


# ================================================================
# 4. HELPER FUNCTIONS
# ================================================================

def show_digit(pattern, title="Digit"):
    plt.imshow(
        np.asarray(pattern).reshape(8, 8),
        cmap="gray_r", vmin=-1, vmax=1,
        interpolation="nearest"
    )
    plt.title(title)
    plt.axis("off")


def pixel_accuracy(original, reconstructed):
    return np.mean(
        np.asarray(original) == np.asarray(reconstructed)
    ) * 100


def get_test_pattern(digit):
    patterns = X_test[y_test == digit]

    if len(patterns) > 0:
        return patterns[0].copy()

    return X_stored[digit].copy()


# ================================================================
# OPTION 1: STORE DIGIT-CODE ASSOCIATIONS
# ================================================================

def option_1():
    print("\n1. STORE DIGIT-CODE PATTERN ASSOCIATIONS")

    table = pd.DataFrame({
        "Digit": range(10),
        "Bipolar Code": [str(c.tolist()) for c in Y_stored]
    })

    display(table)

    print("Forward weight matrix W:")
    print(network.W)

    print("\nTranspose weight matrix W.T:")
    print(network.WT)

    fig, axes = plt.subplots(2, 5, figsize=(10, 5))

    for d, ax in enumerate(axes.flat):
        ax.imshow(X_stored[d].reshape(8, 8), cmap="gray_r")
        ax.set_title(f"Digit {d}")
        ax.axis("off")

    plt.tight_layout()
    plt.show()


# ================================================================
# OPTION 2: FORWARD DIGIT RECALL
# ================================================================

def option_2():
    digit = int(input("Enter digit (0-9): "))

    if digit not in range(10):
        print("Invalid digit!")
        return

    pattern = get_test_pattern(digit)

    recalled_pattern, recalled_code, updates = (
        network.recall_forward(pattern, return_updates=True)
    )

    predicted = code_to_digit(recalled_code)

    print("Expected digit:", digit)
    print("Initial Y-layer state:", network.forward_once(pattern))
    print("Stable output code:", recalled_code)
    print("Identified digit:", predicted)
    print("State updates:", updates)

    plt.figure(figsize=(6, 3))

    plt.subplot(1, 2, 1)
    show_digit(pattern, "Input Digit")

    plt.subplot(1, 2, 2)
    show_digit(recalled_pattern, "Recalled Digit")

    plt.tight_layout()
    plt.show()


# ================================================================
# OPTION 3: REVERSE DIGIT RECONSTRUCTION
# ================================================================

def option_3():
    digit = int(input("Enter digit code to reconstruct (0-9): "))

    if digit not in range(10):
        print("Invalid digit!")
        return

    code = digit_to_code(digit)

    recalled_code, reconstructed = network.recall_backward(code)

    print("Input code:", code)
    print("Recalled code:", recalled_code)
    print("Pixel reconstruction accuracy:",
          round(pixel_accuracy(X_stored[digit], reconstructed), 2), "%")

    plt.figure(figsize=(6, 3))

    plt.subplot(1, 2, 1)
    show_digit(X_stored[digit], "Stored Pattern")

    plt.subplot(1, 2, 2)
    show_digit(reconstructed, "Reconstructed Pattern")

    plt.tight_layout()
    plt.show()


# ================================================================
# OPTION 4: NOISY PATTERN ASSOCIATION TEST
# ================================================================

def option_4():
    digit = int(input("Enter digit (0-9): "))

    if digit not in range(10):
        print("Invalid digit!")
        return

    rng = np.random.default_rng(42)
    original = X_stored[digit]
    results = []

    for noise in [5, 10, 20, 30, 40]:
        noisy = original.copy()
        count = max(1, round(64 * noise / 100))

        indices = rng.choice(64, size=count, replace=False)
        noisy[indices] *= -1

        _, code = network.recall_forward(noisy)
        predicted = code_to_digit(code)

        results.append({
            "Noise (%)": noise,
            "Noisy Pattern": noisy.tolist(),
            "Recalled Code": code.tolist(),
            "Predicted Digit": predicted,
            "Correct": predicted == digit
        })

    display(pd.DataFrame(results))

    fig, axes = plt.subplots(1, 5, figsize=(12, 3))

    rng = np.random.default_rng(42)

    for ax, noise in zip(axes, [5, 10, 20, 30, 40]):
        noisy = original.copy()
        count = max(1, round(64 * noise / 100))
        noisy[rng.choice(64, count, replace=False)] *= -1

        ax.imshow(noisy.reshape(8, 8), cmap="gray_r")
        ax.set_title(f"{noise}% noise")
        ax.axis("off")

    plt.tight_layout()
    plt.show()


# ================================================================
# OPTION 5: PARTIAL DIGIT COMPLETION
# ================================================================

def option_5():
    digit = int(input("Enter digit (0-9): "))

    if digit not in range(10):
        print("Invalid digit!")
        return

    original = X_stored[digit].copy()
    incomplete = original.reshape(8, 8).copy()

    print("1. Remove top rows")
    print("2. Remove middle columns")
    print("3. Remove center rectangle")

    choice = input("Select region (1-3): ")

    if choice == "1":
        incomplete[:3, :] = -1
    elif choice == "2":
        incomplete[:, 2:6] = -1
    elif choice == "3":
        incomplete[2:6, 2:6] = -1
    else:
        print("Invalid choice!")
        return

    incomplete = incomplete.flatten()

    reconstructed, code = network.recall_forward(incomplete)

    print("Recalled digit:", code_to_digit(code))
    print("Reconstruction accuracy:",
          round(pixel_accuracy(original, reconstructed), 2), "%")

    plt.figure(figsize=(9, 3))

    plt.subplot(1, 3, 1)
    show_digit(original, "Original")

    plt.subplot(1, 3, 2)
    show_digit(incomplete, "Incomplete")

    plt.subplot(1, 3, 3)
    show_digit(reconstructed, "Reconstructed")

    plt.tight_layout()
    plt.show()


# ================================================================
# OPTION 6: BIDIRECTIONAL RECALL CONSISTENCY
# ================================================================

def option_6():
    results = []

    for digit in range(10):
        _, code = network.recall_forward(X_stored[digit])
        _, reconstructed = network.recall_backward(code)

        results.append({
            "Digit": digit,
            "Code Correct": np.array_equal(code, Y_stored[digit]),
            "Pixel Accuracy (%)": round(
                pixel_accuracy(X_stored[digit], reconstructed), 2
            ),
            "Exact Reconstruction": np.array_equal(
                X_stored[digit], reconstructed
            )
        })

    df = pd.DataFrame(results)
    display(df)

    print("Exact reconstruction consistency:",
          round(df["Exact Reconstruction"].mean() * 100, 2), "%")

    print("Mean pixel reconstruction accuracy:",
          round(df["Pixel Accuracy (%)"].mean(), 2), "%")


# ================================================================
# OPTION 7: ASSOCIATION CAPACITY CHALLENGE
# ================================================================

def option_7():
    results = []

    for count in [2, 4, 6, 8, 10]:
        bam = BAMNetwork(X_stored[:count], Y_stored[:count])

        forward_correct = 0
        backward_correct = 0
        pixel_scores = []
        consistency = 0

        for digit in range(count):
            _, code = bam.recall_forward(X_stored[digit])

            forward_correct += int(
                np.array_equal(code, Y_stored[digit])
            )

            _, reconstructed = bam.recall_backward(Y_stored[digit])

            backward_correct += int(
                np.array_equal(reconstructed, X_stored[digit])
            )

            pixel_scores.append(
                pixel_accuracy(X_stored[digit], reconstructed)
            )

            _, roundtrip_code = bam.recall_forward(reconstructed)

            consistency += int(
                np.array_equal(roundtrip_code, Y_stored[digit])
            )

        results.append({
            "Associations": count,
            "Forward Accuracy (%)": 100 * forward_correct / count,
            "Backward Exact Accuracy (%)": 100 * backward_correct / count,
            "Pixel Accuracy (%)": np.mean(pixel_scores),
            "Bidirectional Consistency (%)": 100 * consistency / count
        })

    df = pd.DataFrame(results)
    display(df)

    plt.figure(figsize=(8, 4))

    plt.plot(df["Associations"], df["Forward Accuracy (%)"],
             marker="o", label="Forward")

    plt.plot(df["Associations"], df["Backward Exact Accuracy (%)"],
             marker="s", label="Backward exact")

    plt.plot(df["Associations"], df["Bidirectional Consistency (%)"],
             marker="^", label="Consistency")

    plt.xlabel("Number of Associations")
    plt.ylabel("Accuracy (%)")
    plt.title("BAM Association Capacity Analysis")
    plt.ylim(0, 105)
    plt.grid(True)
    plt.legend()
    plt.show()


# ================================================================
# OPTION 8: CROSS-ASSOCIATION ERROR DETECTOR
# ================================================================

def option_8():
    error_matrix = np.zeros((10, 10), dtype=int)
    results = []

    for digit in range(10):
        pattern = get_test_pattern(digit)

        _, code, updates = network.recall_forward(
            pattern, return_updates=True
        )

        predicted = code_to_digit(code)

        if predicted is not None:
            error_matrix[digit, predicted] += 1

        results.append({
            "Input Digit": digit,
            "Expected Code": digit_to_code(digit).tolist(),
            "Recalled Code": code.tolist(),
            "Incorrectly Associated Digit": (
                predicted if predicted != digit else "-"
            ),
            "State Updates": updates
        })

    display(pd.DataFrame(results))

    print("Digit-to-digit association error matrix:")
    display(pd.DataFrame(
        error_matrix,
        index=range(10),
        columns=range(10)
    ))

    off_diagonal = error_matrix.copy()
    np.fill_diagonal(off_diagonal, 0)

    if off_diagonal.sum() > 0:
        actual, predicted = np.unravel_index(
            np.argmax(off_diagonal), off_diagonal.shape
        )
        print(f"Most confused pair: {actual} -> {predicted}")
    else:
        print("No cross-association errors in these test samples.")

    plt.figure(figsize=(7, 5))
    plt.imshow(error_matrix, cmap="Blues")
    plt.colorbar(label="Count")
    plt.xlabel("Predicted Digit")
    plt.ylabel("Actual Digit")
    plt.title("Digit Association Error Matrix")
    plt.xticks(range(10))
    plt.yticks(range(10))
    plt.show()


# ================================================================
# OPTION 9: BAM ASSOCIATIVE MEMORY REPORT
# ================================================================

def option_9():
    forward_correct = 0
    backward_correct = 0
    consistency_correct = 0
    pixel_scores = []

    for digit in range(10):
        _, code = network.recall_forward(X_stored[digit])

        forward_correct += int(
            np.array_equal(code, Y_stored[digit])
        )

        _, reconstructed = network.recall_backward(Y_stored[digit])

        backward_correct += int(
            np.array_equal(reconstructed, X_stored[digit])
        )

        pixel_scores.append(
            pixel_accuracy(X_stored[digit], reconstructed)
        )

        _, roundtrip_code = network.recall_forward(reconstructed)

        consistency_correct += int(
            np.array_equal(roundtrip_code, Y_stored[digit])
        )

    # Reproducible noise experiment on held-out patterns
    rng = np.random.default_rng(2026)
    noise_scores = []

    for digit in range(10):
        original = get_test_pattern(digit)
        highest = 0

        for noise in [5, 10, 20, 30, 40]:
            noisy = original.copy()
            count = max(1, round(64 * noise / 100))
            noisy[rng.choice(64, count, replace=False)] *= -1

            _, code = network.recall_forward(noisy)

            if code_to_digit(code) == digit:
                highest = noise

        noise_scores.append(highest)

    cross_errors = 0

    for digit in range(10):
        _, code = network.recall_forward(get_test_pattern(digit))
        cross_errors += int(code_to_digit(code) != digit)

    report = pd.DataFrame({
        "Metric": [
            "Stored associations",
            "Forward recall accuracy (%)",
            "Backward exact reconstruction (%)",
            "Average pixel reconstruction accuracy (%)",
            "Bidirectional consistency (%)",
            "Mean maximum tested noise tolerated (%)",
            "Held-out cross-association errors"
        ],
        "Result": [
            10,
            round(100 * forward_correct / 10, 2),
            round(100 * backward_correct / 10, 2),
            round(np.mean(pixel_scores), 2),
            round(100 * consistency_correct / 10, 2),
            round(np.mean(noise_scores), 2),
            cross_errors
        ]
    })

    display(report)

    print("Noise tolerance by digit:")

    display(pd.DataFrame({
        "Digit": range(10),
        "Maximum Tested Noise Correctly Recalled (%)": noise_scores
    }))


# ================================================================
# OPTION 10: MENU
# ================================================================

def show_menu():
    actions = {
        "1": option_1,
        "2": option_2,
        "3": option_3,
        "4": option_4,
        "5": option_5,
        "6": option_6,
        "7": option_7,
        "8": option_8,
        "9": option_9
    }

    while True:
        print("\n" + "=" * 65)
        print("BAM HANDWRITTEN DIGIT AND CODE ASSOCIATION SYSTEM")
        print("=" * 65)

        print("1. Store Digit-Code Pattern Associations")
        print("2. Forward Digit Recall Challenge")
        print("3. Reverse Digit Reconstruction Challenge")
        print("4. Noisy Pattern Association Test")
        print("5. Partial Digit Completion Challenge")
        print("6. Bidirectional Recall Consistency Analyzer")
        print("7. Association Capacity Challenge")
        print("8. Cross-Association Error Detector")
        print("9. BAM Associative Memory Report")
        print("10. Exit")

        choice = input("Enter your choice (1-10): ").strip()

        if choice == "10":
            print("Thank you for using BAM!")
            print("Program terminated.")
            break

        elif choice in actions:
            try:
                actions[choice]()
            except ValueError:
                print("Invalid input. Please enter a valid number.")

        else:
            print("Invalid choice. Enter a number from 1 to 10.")


# ================================================================
# START PROGRAM
# ================================================================

print("Libraries imported successfully!")
print("Dataset loaded successfully!")
print("Number of samples:", len(X_all))
print("Image size: 8 x 8")
print("Bipolar input size:", X_stored.shape[1])
print("Bipolar code size:", Y_stored.shape[1])
print("Training associations:", len(X_stored))
print("Forward weight matrix:", network.W.shape)
print("Backward weight matrix:", network.WT.shape)
print("BAM network initialized successfully!")

show_menu()
```
