```python
# ================================================================
# Exercise 7: Implement Boltzmann and Cauchy Machines for Stochastic
# Feature Selection and Energy Optimization on the Wine Quality Dataset
# ================================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from functools import lru_cache

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.dummy import DummyClassifier

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

# ------------------------- Dataset Loading -------------------------

def load_wine_quality_dataset():
    """
    Load red and white Wine Quality datasets from UCI.
    If internet access is unavailable, try local CSV files.
    """
    urls = {
        "red": "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv",
        "white": "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-white.csv",
    }

    frames = []
    errors = []
    for wine_type, url in urls.items():
        try:
            frame = pd.read_csv(url, sep=";")
            frame["wine_type"] = 0 if wine_type == "red" else 1
            frames.append(frame)
        except Exception as exc:
            errors.append(f"{wine_type} dataset URL: {exc}")
            local_name = f"winequality-{wine_type}.csv"
            try:
                frame = pd.read_csv(local_name, sep=";")
                frame["wine_type"] = 0 if wine_type == "red" else 1
                frames.append(frame)
            except Exception as local_exc:
                errors.append(f"{local_name}: {local_exc}")

    if not frames:
        raise RuntimeError(
            "Could not load the Wine Quality dataset. Connect to the internet "
            "or place winequality-red.csv and/or winequality-white.csv in the "
            "notebook's working directory.\n" + "\n".join(errors)
        )

    data = pd.concat(frames, ignore_index=True)
    data = data.dropna().drop_duplicates().reset_index(drop=True)

    # Three-class classification target derived from original quality score:
    # Low (<=5), Medium (=6), High (>=7)
    data["quality_class"] = pd.cut(
        data["quality"],
        bins=[-np.inf, 5, 6, np.inf],
        labels=["Low", "Medium", "High"]
    )

    # Candidate feature-selection columns. quality itself is excluded to
    # prevent target leakage. wine_type is included as a predictor.
    feature_columns = [
        col for col in data.columns
        if col not in ["quality", "quality_class"]
    ]
    X = data[feature_columns].astype(float)
    y = data["quality_class"].astype(str)

    # A fixed train/test split makes candidate feature subsets comparable.
    # If rare class counts make stratification impossible, use an unstratified split.
    try:
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.25, random_state=RANDOM_SEED, stratify=y
        )
    except ValueError:
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.25, random_state=RANDOM_SEED
        )

    print("Dataset loaded successfully!")
    print(f"Rows used: {len(data):,}")
    print(f"Training rows: {len(X_train):,} | Test rows: {len(X_test):,}")
    print(f"Candidate features ({len(feature_columns)}): {feature_columns}")
    print("Target classes:", sorted(y.unique()))
    print("Class distribution:\n", y.value_counts().sort_index().to_string())

    return data, X_train, X_test, y_train, y_test, feature_columns

DATA, X_TRAIN, X_TEST, Y_TRAIN, Y_TEST, FEATURE_NAMES = load_wine_quality_dataset()
N_FEATURES = len(FEATURE_NAMES)

# --------------------------- Energy Model --------------------------

# Energy = classification error + feature-count penalty.
# Lower energy is better. The penalty discourages selecting every feature.
FEATURE_PENALTY = 0.08

def state_to_features(state):
    return [FEATURE_NAMES[i] for i, bit in enumerate(state) if bit == 1]

def evaluate_state(state):
    """
    Evaluate a binary feature-selection state.
    Results are cached because many states are revisited.
    Accuracy is measured on the same held-out test set for fair comparison.
    """
    state = tuple(int(v) for v in state)
    if len(state) != N_FEATURES:
        raise ValueError(f"State must have exactly {N_FEATURES} bits.")

    selected_indices = [i for i, bit in enumerate(state) if bit == 1]
    feature_count = len(selected_indices)

    if feature_count == 0:
        model = DummyClassifier(strategy="most_frequent")
        model.fit(np.zeros((len(X_TRAIN), 1)), Y_TRAIN)
        predictions = model.predict(np.zeros((len(X_TEST), 1)))
    else:
        Xtr = X_TRAIN.iloc[:, selected_indices]
        Xte = X_TEST.iloc[:, selected_indices]
        model = Pipeline([
            ("scaler", StandardScaler()),
            ("classifier", LogisticRegression(
                max_iter=600, class_weight="balanced", random_state=RANDOM_SEED
            ))
        ])
        model.fit(Xtr, Y_TRAIN)
        predictions = model.predict(Xte)

    accuracy = float(accuracy_score(Y_TEST, predictions))
    error = 1.0 - accuracy
    energy = error + FEATURE_PENALTY * (feature_count / N_FEATURES)
    return {
        "accuracy": accuracy,
        "error": error,
        "energy": float(energy),
        "feature_count": feature_count,
        "selected_features": state_to_features(state),
    }

# Cache evaluation by binary state.
_EVALUATION_CACHE = {}
def evaluate_cached(state):
    key = tuple(int(v) for v in state)
    if key not in _EVALUATION_CACHE:
        _EVALUATION_CACHE[key] = evaluate_state(key)
    return _EVALUATION_CACHE[key]

def random_state(min_features=1):
    state = np.random.randint(0, 2, size=N_FEATURES).astype(int)
    if state.sum() < min_features:
        state[np.random.randint(0, N_FEATURES)] = 1
    return state

def print_state_details(state, heading="Feature State"):
    result = evaluate_cached(state)
    print(f"\n{heading}")
    print("-" * len(heading))
    print("Binary state:", "".join(map(str, state.tolist())))
    print(f"Selected feature count: {result['feature_count']} / {N_FEATURES}")
    print("Selected features:", result["selected_features"] or "None")
    print(f"Classification accuracy: {result['accuracy']:.4f} ({result['accuracy']*100:.2f}%)")
    print(f"Classification error: {result['error']:.4f}")
    print(f"Energy: {result['energy']:.6f}")
    return result

def flip_one_bit(state):
    neighbor = np.array(state, dtype=int).copy()
    idx = np.random.randint(0, N_FEATURES)
    neighbor[idx] = 1 - neighbor[idx]
    return neighbor, idx

# ----------------------- Stochastic Machines -----------------------

def run_boltzmann(initial_state=None, iterations=60, initial_temperature=0.25,
                  cooling=0.95, no_improve_limit=12, verbose=False):
    """Simulated annealing using Boltzmann acceptance probability."""
    current = np.array(initial_state, dtype=int).copy() if initial_state is not None else random_state()
    current_result = evaluate_cached(current)
    best = current.copy()
    best_result = current_result
    temperature = float(initial_temperature)

    records = [{
        "iteration": 0, "state": current.copy(), "energy": current_result["energy"],
        "accepted": True, "delta_energy": 0.0, "acceptance_probability": 1.0,
        "higher_energy_accepted": False, "feature_flipped": None
    }]
    accepted_states = [current.copy()]
    visited = {tuple(current.tolist())}
    best_energies = [best_result["energy"]]
    stale = 0
    local_minima = 0
    local_minima_escaped = 0
    trapped = 0
    escape_iteration = None
    local_min_energy = None

    for iteration in range(1, iterations + 1):
        candidate, flipped_idx = flip_one_bit(current)
        candidate_result = evaluate_cached(candidate)
        delta = candidate_result["energy"] - current_result["energy"]
        probability = 1.0 if delta <= 0 else float(np.exp(-delta / max(temperature, 1e-12)))
        accepted = bool(delta <= 0 or np.random.random() < probability)
        higher_accepted = bool(accepted and delta > 0)

        if accepted:
            current = candidate
            current_result = candidate_result
            accepted_states.append(current.copy())
            visited.add(tuple(current.tolist()))

        if candidate_result["energy"] < best_result["energy"] - 1e-12:
            best = candidate.copy()
            best_result = candidate_result
            stale = 0
            if local_min_energy is not None and escape_iteration is None:
                local_minima_escaped += 1
                escape_iteration = iteration
        else:
            stale += 1

        # Approximate local-minimum detection: no new best for a fixed window.
        if stale == no_improve_limit:
            local_minima += 1
            trapped += no_improve_limit
            local_min_energy = current_result["energy"]
            if verbose:
                print(f"  [BM] Possible local minimum near iteration {iteration}, energy={local_min_energy:.6f}")
        elif stale > no_improve_limit:
            trapped += 1

        records.append({
            "iteration": iteration, "state": current.copy(), "energy": current_result["energy"],
            "energy_candidate": candidate_result["energy"], "accepted": accepted,
            "delta_energy": float(delta), "acceptance_probability": probability,
            "higher_energy_accepted": higher_accepted,
            "feature_flipped": FEATURE_NAMES[flipped_idx]
        })
        best_energies.append(best_result["energy"])
        temperature *= cooling

    return {
        "machine": "Boltzmann Machine",
        "initial_state": np.array(initial_state, dtype=int).copy() if initial_state is not None else records[0]["state"].copy(),
        "final_state": current.copy(),
        "best_state": best.copy(),
        "initial_energy": records[0]["energy"],
        "final_energy": current_result["energy"],
        "best_energy": best_result["energy"],
        "best_result": best_result,
        "final_result": current_result,
        "records": records,
        "accepted_states": accepted_states,
        "unique_states": len(visited),
        "higher_energy_accepted": sum(r.get("higher_energy_accepted", False) for r in records),
        "energy_reductions": sum(r.get("accepted", False) and r.get("delta_energy", 0) < 0 for r in records),
        "local_minima": local_minima,
        "local_minima_escaped": local_minima_escaped,
        "trapped_iterations": trapped,
        "escape_iteration": escape_iteration,
        "best_energies": best_energies,
        "iterations": iterations,
        "repeated_states": max(0, len(accepted_states) - len(set(tuple(s.tolist()) for s in accepted_states))),
    }

def run_cauchy(initial_state=None, iterations=60, initial_scale=1.0,
               cooling=0.95, no_improve_limit=12, verbose=False):
    """
    Cauchy annealing: Cauchy-distributed perturbations determine how many
    bits are flipped, enabling occasional larger jumps through state space.
    """
    current = np.array(initial_state, dtype=int).copy() if initial_state is not None else random_state()
    current_result = evaluate_cached(current)
    best = current.copy()
    best_result = current_result
    scale = float(initial_scale)

    records = [{
        "iteration": 0, "state": current.copy(), "energy": current_result["energy"],
        "accepted": True, "delta_energy": 0.0, "acceptance_probability": 1.0,
        "higher_energy_accepted": False, "perturbation": 0.0, "flipped_features": []
    }]
    accepted_states = [current.copy()]
    visited = {tuple(current.tolist())}
    best_energies = [best_result["energy"]]
    stale = 0
    local_minima = 0
    local_minima_escaped = 0
    trapped = 0
    escape_iteration = None
    local_min_energy = None

    for iteration in range(1, iterations + 1):
        # Standard Cauchy(0, scale) perturbation.
        perturbation = float(np.random.standard_cauchy() * scale)
        jump_size = int(np.clip(1 + abs(perturbation), 1, max(1, N_FEATURES)))
        flip_indices = np.random.choice(N_FEATURES, size=jump_size, replace=False)
        candidate = current.copy()
        candidate[flip_indices] = 1 - candidate[flip_indices]
        candidate_result = evaluate_cached(candidate)
        delta = candidate_result["energy"] - current_result["energy"]

        # Cauchy annealing acceptance rule: use heavy-tailed temperature scale.
        probability = 1.0 if delta <= 0 else float(1.0 / (1.0 + delta / max(scale, 1e-12)))
        accepted = bool(delta <= 0 or np.random.random() < probability)
        higher_accepted = bool(accepted and delta > 0)

        if accepted:
            current = candidate
            current_result = candidate_result
            accepted_states.append(current.copy())
            visited.add(tuple(current.tolist()))

        if candidate_result["energy"] < best_result["energy"] - 1e-12:
            best = candidate.copy()
            best_result = candidate_result
            stale = 0
            if local_min_energy is not None and escape_iteration is None:
                local_minima_escaped += 1
                escape_iteration = iteration
        else:
            stale += 1

        if stale == no_improve_limit:
            local_minima += 1
            trapped += no_improve_limit
            local_min_energy = current_result["energy"]
            if verbose:
                print(f"  [CM] Possible local minimum near iteration {iteration}, energy={local_min_energy:.6f}")
        elif stale > no_improve_limit:
            trapped += 1

        records.append({
            "iteration": iteration, "state": current.copy(), "energy": current_result["energy"],
            "energy_candidate": candidate_result["energy"], "accepted": accepted,
            "delta_energy": float(delta), "acceptance_probability": probability,
            "higher_energy_accepted": higher_accepted,
            "perturbation": perturbation,
            "jump_size": jump_size,
            "flipped_features": [FEATURE_NAMES[i] for i in flip_indices]
        })
        best_energies.append(best_result["energy"])
        scale *= cooling

    return {
        "machine": "Cauchy Machine",
        "initial_state": np.array(initial_state, dtype=int).copy() if initial_state is not None else records[0]["state"].copy(),
        "final_state": current.copy(),
        "best_state": best.copy(),
        "initial_energy": records[0]["energy"],
        "final_energy": current_result["energy"],
        "best_energy": best_result["energy"],
        "best_result": best_result,
        "final_result": current_result,
        "records": records,
        "accepted_states": accepted_states,
        "unique_states": len(visited),
        "higher_energy_accepted": sum(r.get("higher_energy_accepted", False) for r in records),
        "energy_reductions": sum(r.get("accepted", False) and r.get("delta_energy", 0) < 0 for r in records),
        "local_minima": local_minima,
        "local_minima_escaped": local_minima_escaped,
        "trapped_iterations": trapped,
        "escape_iteration": escape_iteration,
        "best_energies": best_energies,
        "iterations": iterations,
        "repeated_states": max(0, len(accepted_states) - len(set(tuple(s.tolist()) for s in accepted_states))),
    }

def print_optimization_summary(result):
    print(f"\n{result['machine']} Summary")
    print("=" * (len(result["machine"]) + 8))
    print(f"Initial energy: {result['initial_energy']:.6f}")
    print(f"Final energy:   {result['final_energy']:.6f}")
    print(f"Best energy:    {result['best_energy']:.6f}")
    print(f"Unique states explored: {result['unique_states']}")
    print(f"Energy reductions: {result['energy_reductions']}")
    print(f"Higher-energy states accepted: {result['higher_energy_accepted']}")
    print(f"Local minima encountered: {result['local_minima']}")
    print(f"Local minima escaped: {result['local_minima_escaped']}")
    print(f"Search iterations: {result['iterations']}")
    print_state_details(result["best_state"], "Best Feature Subset")

def plot_energy_curves(results, title="Best Energy During Optimization"):
    plt.figure(figsize=(9, 4.5))
    for result in results:
        plt.plot(result["best_energies"], label=result["machine"])
    plt.xlabel("Iteration")
    plt.ylabel("Best energy so far")
    plt.title(title)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# ----------------------------- Menu 1 ------------------------------

def menu_1():
    state = random_state()
    print_state_details(state, "Initial Random Feature State")
    return state

# ----------------------------- Menu 2 ------------------------------

def menu_2(initial_state=None):
    if initial_state is None:
        initial_state = random_state()
    print_state_details(initial_state, "Starting State for Boltzmann Machine")
    result = run_boltzmann(initial_state, iterations=60, verbose=True)
    print_optimization_summary(result)
    print("\nLast 10 transition records:")
    for r in result["records"][-10:]:
        print(f"Iter {r['iteration']:>3} | ΔE={r.get('delta_energy', 0):+.5f} | "
              f"P={r.get('acceptance_probability', 1):.4f} | "
              f"{'ACCEPTED' if r.get('accepted') else 'REJECTED'} | E={r['energy']:.5f}")
    plot_energy_curves([result], "Boltzmann Machine Energy Optimization")
    return result

# ----------------------------- Menu 3 ------------------------------

def menu_3(initial_state=None):
    if initial_state is None:
        initial_state = random_state()
    print_state_details(initial_state, "Starting State for Cauchy Machine")
    result = run_cauchy(initial_state, iterations=60, verbose=True)
    print_optimization_summary(result)
    print("\nLast 10 transition records:")
    for r in result["records"][-10:]:
        print(f"Iter {r['iteration']:>3} | perturbation={r.get('perturbation', 0):+.4f} | "
              f"flips={r.get('jump_size', 0)} | ΔE={r.get('delta_energy', 0):+.5f} | "
              f"{'ACCEPTED' if r.get('accepted') else 'REJECTED'} | E={r['energy']:.5f}")
    plot_energy_curves([result], "Cauchy Machine Energy Optimization")
    return result

# ----------------------------- Menu 4 ------------------------------

def menu_4():
    initial = random_state()
    print_state_details(initial, "Shared Initial State")
    bm = run_boltzmann(initial, iterations=60)
    cm = run_cauchy(initial, iterations=60)

    for result in [bm, cm]:
        print(f"\n{result['machine']} Transition Analysis")
        print("-" * 36)
        print("Number of Energy Reductions:", result["energy_reductions"])
        print("Higher-Energy States Accepted:", result["higher_energy_accepted"])
        print("Repeated Accepted States:", result["repeated_states"])
        print("Unique States Explored:", result["unique_states"])
        print("Lowest Energy State:", "".join(map(str, result["best_state"].tolist())))
        print(f"Lowest Energy: {result['best_energy']:.6f}")
        print("Transition sequence (first 20 states):")
        for r in result["records"][:20]:
            print(f"  {r['iteration']:>3}: state={''.join(map(str, r['state'].tolist()))} "
                  f"energy={r['energy']:.5f} accepted={r.get('accepted', False)}")
    plot_energy_curves([bm, cm], "Energy-State Transition Comparison")
    return bm, cm

# ----------------------------- Menu 5 ------------------------------

def menu_5():
    initial = random_state()
    print_state_details(initial, "Same Starting State for Local-Minimum Challenge")
    bm = run_boltzmann(initial, iterations=100, no_improve_limit=10)
    cm = run_cauchy(initial, iterations=100, no_improve_limit=10)

    print("\nLocal Minimum Escape Challenge")
    print("=" * 32)
    for result in [bm, cm]:
        print(f"\n{result['machine']}:")
        print(f"  Best/local energy reached: {result['best_energy']:.6f}")
        print(f"  Trapped iterations (approx.): {result['trapped_iterations']}")
        print(f"  Local minima detected: {result['local_minima']}")
        print(f"  Escape iteration: {result['escape_iteration'] if result['escape_iteration'] is not None else 'No escape recorded'}")
        if result["escape_iteration"] is not None:
            print(f"  Best energy after escape: {result['best_energy']:.6f}")
    escaped = [r for r in [bm, cm] if r["escape_iteration"] is not None]
    if escaped:
        winner = min(escaped, key=lambda r: r["escape_iteration"])
        print(f"\nFirst recorded escape: {winner['machine']} at iteration {winner['escape_iteration']}.")
    else:
        winner = min([bm, cm], key=lambda r: r["best_energy"])
        print(f"\nNo escape was recorded by the heuristic; lower best energy: {winner['machine']}.")
    plot_energy_curves([bm, cm], "Local-Minimum Escape Challenge")
    return bm, cm

# ----------------------------- Menu 6 ------------------------------

def menu_6():
    runs = 20
    iterations = 35
    counts = {"Boltzmann Machine": Counter(), "Cauchy Machine": Counter()}
    run_results = {"Boltzmann Machine": [], "Cauchy Machine": []}
    print(f"Running feature stability analysis: {runs} runs per machine, {iterations} iterations each...")
    for run_idx in range(runs):
        initial = random_state()
        bm = run_boltzmann(initial, iterations=iterations)
        cm = run_cauchy(initial, iterations=iterations)
        for name, result in [("Boltzmann Machine", bm), ("Cauchy Machine", cm)]:
            run_results[name].append(result)
            counts[name].update(result["best_result"]["selected_features"])

    rows = []
    for feature in FEATURE_NAMES:
        bm_count = counts["Boltzmann Machine"][feature]
        cm_count = counts["Cauchy Machine"][feature]
        rows.append({
            "Feature": feature,
            "Boltzmann Count": bm_count,
            "Boltzmann Frequency (%)": 100 * bm_count / runs,
            "Cauchy Count": cm_count,
            "Cauchy Frequency (%)": 100 * cm_count / runs,
        })
    stability = pd.DataFrame(rows)
    stability["Average Frequency (%)"] = (
        stability["Boltzmann Frequency (%)"] + stability["Cauchy Frequency (%)"]
    ) / 2
    stability = stability.sort_values("Average Frequency (%)", ascending=False).reset_index(drop=True)
    print("\nFeature Selection Stability Ranking")
    display(stability.round(2) if "display" in globals() else stability.round(2).to_string(index=False))

    high = stability[stability["Average Frequency (%)"] >= 70]["Feature"].tolist()
    moderate = stability[
        (stability["Average Frequency (%)"] >= 35) &
        (stability["Average Frequency (%)"] < 70)
    ]["Feature"].tolist()
    rare = stability[stability["Average Frequency (%)"] < 35]["Feature"].tolist()

    print("\nHighly Stable Features (>=70% average frequency):", high or "None")
    print("Moderately Stable Features (35-69%):", moderate or "None")
    print("Rarely Selected Features (<35%):", rare or "None")

    stability.set_index("Feature")[[
        "Boltzmann Frequency (%)", "Cauchy Frequency (%)"
    ]].plot(kind="bar", figsize=(12, 5))
    plt.title("Feature Selection Frequency Across 20 Runs")
    plt.ylabel("Selection frequency (%)")
    plt.xticks(rotation=55, ha="right")
    plt.tight_layout()
    plt.show()
    return stability, run_results

# ----------------------------- Menu 7 ------------------------------

def menu_7():
    initial = random_state()
    print_state_details(initial, "Identical Initial State for Optimization Battle")
    bm = run_boltzmann(initial, iterations=80)
    cm = run_cauchy(initial, iterations=80)

    comparison = pd.DataFrame([
        {
            "Machine": r["machine"],
            "Initial Energy": r["initial_energy"],
            "Final Energy": r["final_energy"],
            "Minimum Energy": r["best_energy"],
            "Selected Features": r["best_result"]["feature_count"],
            "Classification Accuracy (%)": r["best_result"]["accuracy"] * 100,
            "Unique States": r["unique_states"],
            "Local Minima": r["local_minima"],
            "Local Minima Escaped": r["local_minima_escaped"],
            "Higher-Energy Acceptances": r["higher_energy_accepted"],
            "Iterations": r["iterations"],
        } for r in [bm, cm]
    ])
    print("\nBoltzmann vs Cauchy Optimization Battle")
    display(comparison.round(4) if "display" in globals() else comparison.round(4).to_string(index=False))

    winner = min([bm, cm], key=lambda r: r["best_energy"])
    print(f"\nBest machine by minimum energy: {winner['machine']}")
    print_state_details(winner["best_state"], "Winning Feature Subset")
    plot_energy_curves([bm, cm], "Boltzmann vs Cauchy Optimization Battle")
    return bm, cm, comparison

# ----------------------------- Menu 8 ------------------------------

def menu_8():
    trials = 8
    iterations = 55
    all_results = []
    print(f"Testing {trials} initial states with both machines ({iterations} iterations each)...")
    for trial in range(1, trials + 1):
        initial = random_state()
        print(f"\nTrial {trial}/{trials}")
        bm = run_boltzmann(initial, iterations=iterations)
        cm = run_cauchy(initial, iterations=iterations)
        all_results.extend([bm, cm])
        print(f"  BM best energy={bm['best_energy']:.6f}, accuracy={bm['best_result']['accuracy']:.4f}")
        print(f"  CM best energy={cm['best_energy']:.6f}, accuracy={cm['best_result']['accuracy']:.4f}")

    overall = min(all_results, key=lambda r: r["best_energy"])
    best_state = overall["best_state"]
    selected = state_to_features(best_state)
    removed = [f for f in FEATURE_NAMES if f not in selected]
    result = overall["best_result"]

    print("\nOVERALL BEST FEATURE SUBSET")
    print("===========================")
    print("Machine:", overall["machine"])
    print("Selected features:", selected)
    print("Removed features:", removed)
    print(f"Total candidate features: {N_FEATURES}")
    print(f"Selected feature count: {len(selected)}")
    print(f"Classification accuracy: {result['accuracy']:.4f} ({result['accuracy']*100:.2f}%)")
    print(f"Energy: {result['energy']:.6f}")
    print("Binary state:", "".join(map(str, best_state.tolist())))
    return overall, all_results

# ----------------------------- Menu 9 ------------------------------

def menu_9():
    print("Generating comparative stochastic optimization report...")
    initial = random_state()
    bm = run_boltzmann(initial, iterations=60)
    cm = run_cauchy(initial, iterations=60)
    results = [bm, cm]

    report = pd.DataFrame([{
        "Machine Type": r["machine"],
        "Initial Energy": r["initial_energy"],
        "Final Energy": r["final_energy"],
        "Minimum Energy": r["best_energy"],
        "Selected Feature Count": r["best_result"]["feature_count"],
        "Classification Accuracy (%)": 100 * r["best_result"]["accuracy"],
        "Unique States Explored": r["unique_states"],
        "Local Minima Encountered": r["local_minima"],
        "Local Minima Escaped": r["local_minima_escaped"],
        "Higher-Energy States Accepted": r["higher_energy_accepted"],
        "Search Iterations": r["iterations"],
        "Best Feature Subset": ", ".join(r["best_result"]["selected_features"]),
    } for r in results])

    print("\nSTOCHASTIC OPTIMIZATION REPORT")
    print("==============================")
    display(report.round(4) if "display" in globals() else report.round(4).to_string(index=False))

    winner = min(results, key=lambda r: r["best_energy"])
    print(f"\nBest stochastic machine: {winner['machine']}")
    print("Best feature subset:", winner["best_result"]["selected_features"])
    print(f"Minimum energy: {winner['best_energy']:.6f}")
    print(f"Classification accuracy: {winner['best_result']['accuracy']*100:.2f}%")

    # Save the report for later use in the notebook's current directory.
    report.to_csv("exercise7_stochastic_optimization_report.csv", index=False)
    print("Report saved as exercise7_stochastic_optimization_report.csv")
    plot_energy_curves(results, "Final Stochastic Optimization Report")
    return report, winner

# ------------------------------ Menu -------------------------------

def print_menu():
    print("\n" + "=" * 72)
    print("BOLTZMANN AND CAUCHY MACHINE FEATURE SELECTION SYSTEM")
    print("Wine Quality Dataset | Stochastic Feature Selection and Energy Optimization")
    print("=" * 72)
    print("1. Generate Initial Feature State")
    print("2. Boltzmann Machine Feature Selection")
    print("3. Cauchy Machine Feature Selection")
    print("4. Energy State Transition Analyzer")
    print("5. Local Minimum Escape Challenge")
    print("6. Feature Stability Analyzer")
    print("7. Boltzmann vs Cauchy Optimization Battle")
    print("8. Best Feature Subset Finder")
    print("9. Stochastic Optimization Report")
    print("10. Exit")
    print("-" * 72)

def main():
    saved_initial_state = None
    last_bm = None
    last_cm = None

    while True:
        print_menu()
        choice = input("Enter your choice (1-10): ").strip()

        try:
            if choice == "1":
                saved_initial_state = menu_1()
            elif choice == "2":
                last_bm = menu_2(saved_initial_state)
            elif choice == "3":
                last_cm = menu_3(saved_initial_state)
            elif choice == "4":
                last_bm, last_cm = menu_4()
            elif choice == "5":
                last_bm, last_cm = menu_5()
            elif choice == "6":
                menu_6()
            elif choice == "7":
                last_bm, last_cm, _ = menu_7()
            elif choice == "8":
                menu_8()
            elif choice == "9":
                menu_9()
            elif choice == "10":
                print("\nThank you for using the Boltzmann and Cauchy Machine System!")
                print("Program terminated.")
                break
            else:
                print("Invalid choice. Please enter a number from 1 to 10.")
        except KeyboardInterrupt:
            print("\nOperation interrupted by user. Returning to the menu.")
        except Exception as exc:
            print(f"\nAn error occurred: {exc}")
            print("Please check dataset availability and try again.")

# Run the menu-driven application.
main()
```
